```python
#!/usr/bin/env python3
"""
================================================================================
ASL ALPHABET CLASSIFICATION - UNIFIED TRAINING PIPELINE
================================================================================
Complete training pipeline combining best practices from both training approaches.

FEATURES:
1. ✅ Comprehensive dataset analysis and metrics
2. ✅ MediaPipe hand cropping with GPU optimization
3. ✅ Model-specific preprocessing (matches inference exactly)
4. ✅ Resume training from existing models OR train from scratch
5. ✅ GPU-optimized training with minimal CPU usage
6. ✅ All metrics saved and printed to terminal
7. ✅ Training graphs and visualizations
8. ✅ Confusion matrices and classification reports

USAGE:
    python train.py

CONFIGURATION:
    Edit CONFIG dict below to customize paths, models, and training parameters.
================================================================================
"""

# ==================== INSTALLATION CHECK ====================
import subprocess
import sys

def install_if_missing(package, pip_name=None):
    """Install package if not available"""
    try:
        __import__(package)
    except ImportError:
        pip_name = pip_name or package
        print(f"Installing {pip_name}...")
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', pip_name, '-q'])

# Check and install dependencies
install_if_missing('mediapipe')
install_if_missing('cv2', 'opencv-python')
install_if_missing('albumentations')
install_if_missing('sklearn', 'scikit-learn')
install_if_missing('tqdm')
install_if_missing('plotly')
install_if_missing('kaleido')  # For saving plotly figures

# ==================== IMPORTS ====================
import os
import shutil
import numpy as np
import pandas as pd
import matplotlib
# Check if running in notebook (Kaggle/Jupyter) - use inline display
import sys
IN_NOTEBOOK = 'ipykernel' in sys.modules or 'IPython' in sys.modules
if not IN_NOTEBOOK:
    matplotlib.use('Agg')  # Non-interactive backend only if not in notebook
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import seaborn as sns
from pathlib import Path
import cv2
import mediapipe as mp
from tqdm import tqdm
import json

# Plotly for interactive 3D visualizations
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio

# Set Plotly renderer for Kaggle/Jupyter notebooks
if 'ipykernel' in sys.modules or 'IPython' in sys.modules:
    pio.renderers.default = 'notebook'  # Best for Kaggle

# For inline display in Kaggle notebooks
if IN_NOTEBOOK:
    try:
        from IPython.display import display, Image, HTML, clear_output
        from IPython import get_ipython
        get_ipython().run_line_magic('matplotlib', 'inline')
        print("📊 Notebook detected - plots will display inline!")
    except:
        pass
from collections import Counter
from datetime import datetime
import gc
import time
import warnings
warnings.filterwarnings('ignore')

# TensorFlow imports with GPU configuration
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.applications import ResNet50, EfficientNetV2B3, InceptionV3
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet_preprocess
from tensorflow.keras.applications.efficientnet_v2 import preprocess_input as effnet_preprocess
from tensorflow.keras.applications.inception_v3 import preprocess_input as inception_preprocess
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau, TensorBoard
import albumentations as A

# Sklearn imports
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    precision_score, recall_score, f1_score, top_k_accuracy_score,
    balanced_accuracy_score, cohen_kappa_score, matthews_corrcoef,
    roc_auc_score, log_loss, roc_curve, auc, precision_recall_curve,
    average_precision_score
)
from sklearn.preprocessing import label_binarize

# Set seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# ==================== GPU CONFIGURATION ====================
def configure_gpu():
    """Configure GPU for optimal performance, minimize CPU usage"""
    print("\n" + "=" * 70)
    print("GPU CONFIGURATION")
    print("=" * 70)
    
    # Force single GPU usage
    os.environ["CUDA_VISIBLE_DEVICES"] = "0"
    os.environ["TF_FORCE_GPU_ALLOW_GROWTH"] = "true"
    
    # Optimize TensorFlow for GPU
    os.environ["TF_GPU_THREAD_MODE"] = "gpu_private"
    os.environ["TF_GPU_THREAD_COUNT"] = "2"
    
    # Reduce CPU parallelism to minimize CPU usage
    tf.config.threading.set_inter_op_parallelism_threads(2)
    tf.config.threading.set_intra_op_parallelism_threads(2)
    
    gpus = tf.config.list_physical_devices('GPU')
    
    if gpus:
        try:
            # Use only first GPU
            tf.config.set_visible_devices(gpus[0], 'GPU')
            # Enable memory growth to avoid OOM
            tf.config.experimental.set_memory_growth(gpus[0], True)
            
            # Set memory limit if needed (uncomment and adjust)
            # tf.config.set_logical_device_configuration(
            #     gpus[0],
            #     [tf.config.LogicalDeviceConfiguration(memory_limit=8192)]
            # )
            
            print(f"✅ GPU detected: {gpus[0].name}")
            print(f"✅ Memory growth enabled")
            print(f"✅ Using GPU 0 only")
            
            # Get GPU details
            try:
                gpu_details = tf.config.experimental.get_device_details(gpus[0])
                if gpu_details:
                    print(f"✅ GPU Details: {gpu_details}")
            except:
                pass
                
            return True
            
        except RuntimeError as e:
            print(f"⚠️ GPU configuration error: {e}")
            return False
    else:
        print("⚠️ No GPU detected - using CPU (training will be slower)")
        return False

# Configure GPU at import time
HAS_GPU = configure_gpu()

print(f"\n📊 TensorFlow Version: {tf.__version__}")
print(f"📊 GPU Available: {HAS_GPU}")
print(f"📊 Physical Devices: {tf.config.list_physical_devices()}")

# ==================== CONFIGURATION ====================
CONFIG = {
    # ===== PATHS (KAGGLE ENVIRONMENT) =====
    # Dataset paths for Kaggle
    'train_path': '/kaggle/input/aslamerican-sign-language-aplhabet-dataset/ASL_Alphabet_Dataset/asl_alphabet_train',  # Raw training images
    'test_path': '/kaggle/input/aslamerican-sign-language-aplhabet-dataset/ASL_Alphabet_Dataset/asl_alphabet_test',    # Raw test images
    'cropped_train_path': '/kaggle/working/cropped_train',    # MediaPipe cropped training
    'cropped_test_path': '/kaggle/working/cropped_test',      # MediaPipe cropped test
    'models_path': '/kaggle/working/models',                  # Saved models
    'metrics_path': '/kaggle/working/metrics',                # Saved metrics and graphs
    
    # ===== MODEL PATHS FOR RESUME =====
    # Set these to your existing model paths for resume training
    # If None or file doesn't exist, will train from scratch
    'existing_models': {
        'ResNet50': '/kaggle/input/resnet50-best-1-keras/keras/1/2',           # Change to your model input path
        'InceptionV3': '/kaggle/input/inceptionv3-best-keras/keras/1/1/InceptionV3_best.keras',     # Change to your model input path
        'EfficientNetV2B3': '/kaggle/input/efficientnetv2b3-best-1-keras/keras/1/1/EfficientNetV2B3_best.keras'  # Change to your model input path
    },
    
    # ===== DATASET CONFIG =====
    'skip_folders': ['del', 'nothing', 'space'],
    'skip_test_files': ['nothing_test.jpg', 'del_test.jpg', 'space_test.jpg'],
    'images_per_class': 1000,  # Set to None for ALL images
    'test_images_per_class_from_train': 100,
    'require_mediapipe_detection': True,
    'use_cached_crops': True,  # Skip MediaPipe if crops exist
    
    # ===== TRAINING CONFIG =====
    'img_size': (224, 224),
    'batch_size': 32,
    'epochs': 35,
    'val_split': 0.15,
    'resume_epochs': 15,       # Additional epochs when resuming
    'fine_tune_epochs': 5,     # Few epochs for fine-tuning already-good models
    
    # ===== MODELS TO TRAIN =====
    # Only these models will be trained (in order)
    # Include ALL models you want to train - they will be trained in this order
    'train_models': ['EfficientNetV2B3', 'ResNet50', 'InceptionV3'],
    
    # ===== MEDIAPIPE CONFIG =====
    'mediapipe_confidence': 0.6,
    'mediapipe_margin': 30,
    'mediapipe_model_complexity': 1,  # 0=lite, 1=full
}

# Create directories
for path_key in ['cropped_train_path', 'cropped_test_path', 'models_path', 'metrics_path']:
    os.makedirs(CONFIG[path_key], exist_ok=True)

# ==================== LOGGING UTILITIES ====================
class Logger:
    """Simple logger that prints to terminal and saves to file"""
    
    def __init__(self, log_file=None):
        self.log_file = log_file or os.path.join(CONFIG['metrics_path'], 'training_log.txt')
        self.start_time = datetime.now()
        
        # Clear previous log
        with open(self.log_file, 'w') as f:
            f.write(f"ASL Training Log - {self.start_time.strftime('%Y-%m-%d %H:%M:%S')}\n")
            f.write("=" * 70 + "\n\n")
    
    def log(self, message, also_print=True):
        """Log message to file and optionally print"""
        timestamp = datetime.now().strftime('%H:%M:%S')
        formatted = f"[{timestamp}] {message}"
        
        if also_print:
            print(message)
        
        with open(self.log_file, 'a') as f:
            f.write(formatted + "\n")
    
    def section(self, title):
        """Print section header"""
        header = "\n" + "=" * 70 + f"\n{title}\n" + "=" * 70 + "\n"
        self.log(header)

logger = Logger()

# ==================== VISUALIZATION HELPER ====================
def save_and_show_plot(save_path, title=None):
    """Save matplotlib plot to file AND display inline in notebook"""
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    logger.log(f"📊 Saved: {save_path}")
    plt.show()  # Always show - works in terminal and notebook
    plt.close()

def save_and_show_plotly(fig, save_path, title=None):
    """Save Plotly figure to file AND display ALWAYS (terminal/notebook)
    
    This creates INTERACTIVE 3D plots that you can rotate, zoom, pan!
    ALL plots are shown in browser for interactivity.
    """
    # Save as HTML (interactive)
    html_path = save_path.replace('.png', '.html')
    fig.write_html(html_path, auto_open=False)
    logger.log(f"📊 Interactive HTML saved: {html_path}")
    
    # Also save as static image for reference (optional)
    try:
        fig.write_image(save_path, width=1200, height=800, scale=2)
        logger.log(f"📊 Static image saved: {save_path}")
    except Exception as e:
        pass  # kaleido may not be available
    
    # ALWAYS display the plot - opens in browser or shows inline
    try:
        if IN_NOTEBOOK:
            # Notebook display
            from IPython.display import display, HTML
            html_str = fig.to_html(full_html=False, include_plotlyjs='cdn')
            display(HTML(html_str))
            logger.log(f"📊 Displayed interactive plot: {title or 'Plot'}")
        else:
            # Terminal/script - open in browser
            fig.show()
            logger.log(f"📊 Opened interactive plot in browser: {title or 'Plot'}")
    except Exception as e:
        # Final fallback - just show
        try:
            fig.show()
        except:
            logger.log(f"⚠️ Could not display plot: {e}")

def display_image_file(image_path, title=None):
    """Display a saved image file inline in notebook"""
    if IN_NOTEBOOK and os.path.exists(image_path):
        try:
            from IPython.display import display, Image
            if title:
                print(f"\n📊 {title}")
            display(Image(filename=image_path))
        except:
            pass

# ==================== 3D BAR HELPER FUNCTION ====================
def create_3d_bar_mesh(x_center, y_center, z_height, bar_width=0.4, bar_depth=0.4, color='blue', 
                       opacity=1.0, name='', hovertemplate=''):
    """
    Create a solid 3D bar using Mesh3d with proper triangular faces.
    
    This creates a proper rectangular prism (cuboid) with all 6 faces rendered correctly.
    Each face is made of 2 triangles = 12 triangles total for a closed solid bar.
    
    Args:
        x_center: X position of bar center
        y_center: Y position of bar center  
        z_height: Height of bar (from 0)
        bar_width: Width in X direction
        bar_depth: Depth in Y direction
        color: Bar color
        opacity: Bar opacity (0-1)
        name: Name for legend
        hovertemplate: Hover text template
    
    Returns:
        go.Mesh3d trace for the bar
    """
    # Half dimensions
    hw = bar_width / 2  # half width
    hd = bar_depth / 2  # half depth
    
    # 8 vertices of the cuboid (bar)
    # Bottom face (z=0): vertices 0,1,2,3
    # Top face (z=height): vertices 4,5,6,7
    vertices_x = [
        x_center - hw, x_center + hw, x_center + hw, x_center - hw,  # bottom: 0,1,2,3
        x_center - hw, x_center + hw, x_center + hw, x_center - hw   # top: 4,5,6,7
    ]
    vertices_y = [
        y_center - hd, y_center - hd, y_center + hd, y_center + hd,  # bottom
        y_center - hd, y_center - hd, y_center + hd, y_center + hd   # top
    ]
    vertices_z = [
        0, 0, 0, 0,           # bottom face at z=0
        z_height, z_height, z_height, z_height  # top face at z=height
    ]
    
    # 12 triangular faces (2 per face × 6 faces)
    # Each face needs 2 triangles defined by vertex indices (i, j, k)
    # The vertices must be in counter-clockwise order when viewed from outside
    
    # Bottom face (z=0): vertices 0,1,2,3 - looking from below
    # Top face (z=h): vertices 4,5,6,7 - looking from above
    # Front face (y=-): vertices 0,1,5,4
    # Back face (y=+): vertices 2,3,7,6
    # Left face (x=-): vertices 0,3,7,4
    # Right face (x=+): vertices 1,2,6,5
    
    i_faces = [
        0, 0,  # bottom: triangles (0,1,2) and (0,2,3)
        4, 4,  # top: triangles (4,6,5) and (4,7,6)
        0, 0,  # front: triangles (0,5,1) and (0,4,5)
        2, 2,  # back: triangles (2,7,3) and (2,6,7)
        0, 0,  # left: triangles (0,3,7) and (0,7,4)
        1, 1,  # right: triangles (1,5,6) and (1,6,2)
    ]
    j_faces = [
        1, 2,  # bottom
        6, 7,  # top
        5, 4,  # front
        7, 6,  # back
        3, 7,  # left
        5, 6,  # right
    ]
    k_faces = [
        2, 3,  # bottom
        5, 6,  # top
        1, 5,  # front
        3, 7,  # back
        7, 4,  # left
        6, 2,  # right
    ]
    
    return go.Mesh3d(
        x=vertices_x,
        y=vertices_y,
        z=vertices_z,
        i=i_faces,
        j=j_faces,
        k=k_faces,
        color=color,
        opacity=opacity,
        name=name,
        showlegend=False,
        hovertemplate=hovertemplate,
        flatshading=True,  # Better solid appearance
        lighting=dict(
            ambient=0.7,
            diffuse=0.8,
            specular=0.2,
            roughness=0.5,
            fresnel=0.1
        ),
        lightposition=dict(x=100, y=200, z=300)
    )

# ==================== SAVEDMODEL LOADER ====================
def load_model_any_format(model_path, model_name=None):
    """
    Load a model from any format (Keras 3 .keras, H5, or TensorFlow SavedModel).
    
    Args:
        model_path: Path to the model file/directory
        model_name: Optional model name for logging
    
    Returns:
        (model, format_type) tuple or (None, error_message)
    """
    name = model_name or os.path.basename(model_path)
    
    # Helper function to find model files in a directory
    def find_model_file(base_path):
        """Search for model files in directory tree"""
        model_extensions = ['.keras', '.h5', '.hdf5']
        savedmodel_markers = ['saved_model.pb', 'saved_model.pbtxt']
        
        if not os.path.exists(base_path):
            return None, None
        
        # If it's a file, return it directly
        if os.path.isfile(base_path):
            return base_path, 'file'
        
        # Check if this directory is a SavedModel
        for marker in savedmodel_markers:
            if os.path.exists(os.path.join(base_path, marker)):
                return base_path, 'savedmodel'
        
        # Search for model files recursively (max depth 3)
        for root, dirs, files in os.walk(base_path):
            depth = root[len(base_path):].count(os.sep)
            if depth > 3:
                continue
            
            # Check for SavedModel markers in subdirs
            for marker in savedmodel_markers:
                if marker in files:
                    return root, 'savedmodel'
            
            # Check for keras/h5 files
            for f in files:
                for ext in model_extensions:
                    if f.endswith(ext):
                        return os.path.join(root, f), 'file'
        
        return None, None
    
    # First, try to find the actual model file/directory
    actual_path, path_type = find_model_file(model_path)
    
    if actual_path and actual_path != model_path:
        logger.log(f"   📂 Found model at: {actual_path}")
        model_path = actual_path
    
    # Log directory contents for debugging
    if os.path.isdir(model_path):
        try:
            contents = os.listdir(model_path)
            logger.log(f"   📁 Directory contents: {contents[:10]}{'...' if len(contents) > 10 else ''}")
        except:
            pass
    
    # Try standard Keras load first (.keras or .h5)
    try:
        model = keras.models.load_model(model_path)
        logger.log(f"✅ Loaded {name} using keras.models.load_model()")
        return model, 'keras'
    except Exception as e1:
        keras_error = str(e1)
    
    # If it's a directory, try multiple approaches for SavedModel format
    if os.path.isdir(model_path):
        
        # Check for saved_model.pb first
        has_savedmodel_pb = os.path.exists(os.path.join(model_path, 'saved_model.pb'))
        has_savedmodel_pbtxt = os.path.exists(os.path.join(model_path, 'saved_model.pbtxt'))
        
        if not has_savedmodel_pb and not has_savedmodel_pbtxt:
            # Not a valid SavedModel directory - search subdirectories
            logger.log(f"   ⚠️ No saved_model.pb found, searching subdirectories...")
            
            for subdir in os.listdir(model_path):
                subpath = os.path.join(model_path, subdir)
                if os.path.isdir(subpath):
                    if os.path.exists(os.path.join(subpath, 'saved_model.pb')):
                        logger.log(f"   📂 Found SavedModel in: {subpath}")
                        model_path = subpath
                        has_savedmodel_pb = True
                        break
                    # Check for .keras files
                    for f in os.listdir(subpath):
                        if f.endswith('.keras') or f.endswith('.h5'):
                            keras_file = os.path.join(subpath, f)
                            logger.log(f"   📂 Found Keras file: {keras_file}")
                            try:
                                model = keras.models.load_model(keras_file)
                                logger.log(f"✅ Loaded {name} from {keras_file}")
                                return model, 'keras'
                            except Exception as e:
                                logger.log(f"   Failed to load {keras_file}: {e}")
        
        # Approach 1: Try tf.saved_model.load (TF2 native)
        if has_savedmodel_pb or has_savedmodel_pbtxt:
            try:
                logger.log(f"   Trying tf.saved_model.load()...")
                imported = tf.saved_model.load(model_path)
                
                # Check if it has a keras model signature
                if hasattr(imported, 'signatures'):
                    signatures = list(imported.signatures.keys())
                    logger.log(f"   Found signatures: {signatures}")
                    
                    # Get the serving function
                    if 'serving_default' in signatures:
                        serve_fn = imported.signatures['serving_default']
                        
                        # Create a wrapper model for inference
                        class SavedModelWrapper(keras.Model):
                            def __init__(self, serve_fn, **kwargs):
                                super().__init__(**kwargs)
                                self._serve_fn = serve_fn
                                # Try to get output shape from signature
                                output_info = list(serve_fn.structured_outputs.values())[0]
                                self._output_classes = output_info.shape[-1] if output_info.shape else 26
                            
                            def call(self, inputs):
                                # The serve function expects a dict with input tensor
                                result = self._serve_fn(inputs)
                                # Return the first output value
                                return list(result.values())[0]
                            
                            @property
                            def output_shape(self):
                                return (None, self._output_classes)
                        
                        model = SavedModelWrapper(serve_fn, name=f"{name}_TF2")
                        # Build the model with sample input
                        model.build(input_shape=(None, 224, 224, 3))
                        
                        logger.log(f"✅ Loaded {name} using tf.saved_model.load() + wrapper")
                        logger.log(f"   ⚠️ Note: This is inference-only - cannot be trained further")
                        return model, 'savedmodel'
                        
            except Exception as e2:
                logger.log(f"   tf.saved_model.load failed: {e2}")
        
        # Approach 2: Try TFSMLayer with various endpoints
        try:
            logger.log(f"   Trying TFSMLayer...")
            endpoints_to_try = ['serving_default', 'serve', '__call__', 'call']
            
            for endpoint in endpoints_to_try:
                try:
                    tfsm_layer = keras.layers.TFSMLayer(model_path, call_endpoint=endpoint)
                    
                    # Wrap in a functional model for compatibility
                    inputs = keras.Input(shape=(224, 224, 3))
                    outputs = tfsm_layer(inputs)
                    model = keras.Model(inputs, outputs, name=f"{name}_TFSM")
                    
                    logger.log(f"✅ Loaded {name} using TFSMLayer (endpoint: {endpoint})")
                    logger.log(f"   ⚠️ Note: This is inference-only - cannot be trained further")
                    return model, 'savedmodel'
                except Exception as e:
                    continue
            
            logger.log(f"   TFSMLayer failed with all endpoints: {endpoints_to_try}")
            
        except Exception as e3:
            logger.log(f"   TFSMLayer approach failed: {e3}")
        
        # Approach 3: Check if there's a keras_metadata.pb file (indicates Keras SavedModel)
        keras_metadata_path = os.path.join(model_path, 'keras_metadata.pb')
        if os.path.exists(keras_metadata_path):
            logger.log(f"   Found keras_metadata.pb - this is a Keras 2.x SavedModel")
            logger.log(f"   ⚠️ Keras 3 cannot load Keras 2 SavedModels directly")
            logger.log(f"   💡 Tip: Re-save the model in .keras format using Keras 2.x:")
            logger.log(f"      model = tf.keras.models.load_model('{model_path}')")
            logger.log(f"      model.save('{model_path}.keras')")
            return None, "Keras 2 SavedModel not compatible with Keras 3. Re-save as .keras format."
        
        return None, f"SavedModel load failed with all approaches"
    
    # Neither worked
    return None, f"Cannot load: {keras_error}"

# ==================== DATASET ANALYSIS ====================
def analyze_dataset():
    """Comprehensive dataset analysis with metrics"""
    
    logger.section("DATASET ANALYSIS")
    
    metrics = {
        'total_classes': 0,
        'total_train_images': 0,
        'total_test_images': 0,
        'class_distribution': {},
        'image_sizes': [],
        'analysis_time': None
    }
    
    start_time = time.time()
    
    # Check if paths exist
    if not os.path.exists(CONFIG['train_path']):
        logger.log(f"❌ Training path not found: {CONFIG['train_path']}")
        logger.log("Please update CONFIG['train_path'] to point to your dataset")
        return metrics
    
    # Get all class folders
    all_folders = [d for d in os.listdir(CONFIG['train_path']) 
                   if os.path.isdir(os.path.join(CONFIG['train_path'], d))]
    class_names = sorted([c for c in all_folders if c not in CONFIG['skip_folders']])
    
    metrics['total_classes'] = len(class_names)
    metrics['class_names'] = class_names
    
    logger.log(f"\n📁 Dataset Path: {CONFIG['train_path']}")
    logger.log(f"📊 Total folders found: {len(all_folders)}")
    logger.log(f"🚫 Skipped folders: {CONFIG['skip_folders']}")
    logger.log(f"✅ Valid classes: {len(class_names)}")
    logger.log(f"📝 Classes: {class_names}\n")
    
    # Analyze each class
    logger.log("Class Distribution:")
    logger.log("-" * 50)
    
    total_images = 0
    class_counts = {}
    
    for class_name in class_names:
        class_dir = os.path.join(CONFIG['train_path'], class_name)
        images = [f for f in os.listdir(class_dir) 
                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        count = len(images)
        class_counts[class_name] = count
        total_images += count
        
        # Sample image size
        if images and len(metrics['image_sizes']) < 10:
            sample_path = os.path.join(class_dir, images[0])
            img = cv2.imread(sample_path)
            if img is not None:
                metrics['image_sizes'].append(img.shape[:2])
        
        logger.log(f"  {class_name}: {count:,} images")
    
    metrics['total_train_images'] = total_images
    metrics['class_distribution'] = class_counts
    
    logger.log("-" * 50)
    logger.log(f"\n📊 Total Training Images: {total_images:,}")
    logger.log(f"📊 Average per class: {total_images // len(class_names):,}")
    logger.log(f"📊 Min per class: {min(class_counts.values()):,}")
    logger.log(f"📊 Max per class: {max(class_counts.values()):,}")
    
    if metrics['image_sizes']:
        avg_h = sum(s[0] for s in metrics['image_sizes']) // len(metrics['image_sizes'])
        avg_w = sum(s[1] for s in metrics['image_sizes']) // len(metrics['image_sizes'])
        logger.log(f"📊 Average image size: {avg_h}x{avg_w}")
    
    # Test set analysis
    if os.path.exists(CONFIG['test_path']):
        test_files = [f for f in os.listdir(CONFIG['test_path']) 
                     if f.lower().endswith('.jpg') and '_test' in f]
        valid_test = [f for f in test_files if f not in CONFIG['skip_test_files']]
        metrics['total_test_images'] = len(valid_test)
        logger.log(f"📊 Test images: {len(valid_test)}")
    
    # Sampling info
    if CONFIG['images_per_class']:
        expected_total = CONFIG['images_per_class'] * len(class_names)
        logger.log(f"\n⚙️ Sampling: {CONFIG['images_per_class']} images per class")
        logger.log(f"⚙️ Expected total: {expected_total:,} images")
    else:
        logger.log(f"\n⚙️ Using ALL images (~{total_images:,})")
    
    metrics['analysis_time'] = time.time() - start_time
    logger.log(f"\n⏱️ Analysis time: {metrics['analysis_time']:.2f}s")
    
    # Save metrics
    metrics_file = os.path.join(CONFIG['metrics_path'], 'dataset_metrics.json')
    with open(metrics_file, 'w') as f:
        # Convert to JSON-serializable
        save_metrics = {k: v for k, v in metrics.items() if k != 'image_sizes'}
        json.dump(save_metrics, f, indent=2)
    logger.log(f"💾 Metrics saved to: {metrics_file}")
    
    # Visualize class distribution (single combined visualization)
    visualize_class_distribution(class_counts)
    
    # Show sample images from each class (skip 3D distribution - redundant)
    visualize_dataset_samples(class_names)
    
    return metrics

def visualize_class_distribution(class_counts):
    """Create INTERACTIVE 3D class distribution visualization with SOLID bars"""
    
    logger.log("\n📊 Creating interactive 3D class distribution...")
    
    classes = list(class_counts.keys())
    counts = list(class_counts.values())
    n_classes = len(classes)
    max_count = max(counts)
    
    # Create 3D bar chart using solid Mesh3d bars
    fig = go.Figure()
    
    for i, (cls, count) in enumerate(zip(classes, counts)):
        norm = count / max_count
        
        # Color gradient based on count (blue to purple spectrum)
        r = int(50 + 150 * norm)
        g = int(100 + 50 * (1 - norm))
        b = int(200 - 50 * norm)
        color = f'rgb({r}, {g}, {b})'
        
        # Add solid 3D bar using helper function
        bar = create_3d_bar_mesh(
            x_center=i,
            y_center=0,
            z_height=count,
            bar_width=0.7,
            bar_depth=0.7,
            color=color,
            opacity=0.95,
            name=cls,
            hovertemplate=f'<b>Class {cls}</b><br>Images: {count:,}<extra></extra>'
        )
        fig.add_trace(bar)
    
    # Add text labels at top of bars
    fig.add_trace(go.Scatter3d(
        x=list(range(n_classes)),
        y=[0] * n_classes,
        z=[c + max_count * 0.03 for c in counts],
        mode='text',
        text=classes,
        textfont=dict(size=12, color='black', family='Arial Black'),
        hoverinfo='skip',
        showlegend=False
    ))
    
    fig.update_layout(
        title=dict(
            text='🔤 ASL Dataset - 3D Class Distribution (Rotate to explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(
                title='Class', 
                tickvals=list(range(n_classes)), 
                ticktext=classes, 
                tickangle=45,
                gridcolor='lightgray'
            ),
            yaxis=dict(title='', showticklabels=False, range=[-1, 1], showgrid=False),
            zaxis=dict(title='Number of Images', gridcolor='lightgray'),
            camera=dict(eye=dict(x=1.5, y=1.5, z=0.8)),
            aspectmode='manual',
            aspectratio=dict(x=2, y=0.5, z=1),
            bgcolor='rgba(250,250,250,0.9)'
        ),
        template='plotly_white',
        height=650,
        margin=dict(l=0, r=0, t=60, b=0)
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], 'class_distribution.png')
    save_and_show_plotly(fig, save_path, '3D Class Distribution')

def visualize_class_distribution_3d(class_counts):
    """Create INTERACTIVE 3D surface and bar visualization for class distribution"""
    
    logger.log("\n📊 Creating interactive 3D class distribution...")
    
    classes = list(class_counts.keys())
    counts = list(class_counts.values())
    n_classes = len(classes)
    
    # Create subplots: 3D bars + 3D surface
    fig = make_subplots(
        rows=1, cols=2,
        specs=[[{"type": "scene"}, {"type": "scene"}]],
        subplot_titles=['3D Bar Chart', '3D Surface Heatmap'],
        horizontal_spacing=0.05
    )
    
    # ===== 3D Bar Chart using SOLID bars =====
    max_count = max(counts)
    for i, (cls, count) in enumerate(zip(classes, counts)):
        norm = count / max_count
        
        # Color gradient based on count
        r = int(30 + 200 * norm)
        g = int(150 - 50 * norm)
        b = int(220 - 150 * norm)
        color = f'rgb({r}, {g}, {b})'
        
        # Create solid 3D bar using helper (need to manually add for subplots)
        hw, hd = 0.35, 0.35
        x_c, y_c = i, 0
        
        vertices_x = [x_c-hw, x_c+hw, x_c+hw, x_c-hw, x_c-hw, x_c+hw, x_c+hw, x_c-hw]
        vertices_y = [y_c-hd, y_c-hd, y_c+hd, y_c+hd, y_c-hd, y_c-hd, y_c+hd, y_c+hd]
        vertices_z = [0, 0, 0, 0, count, count, count, count]
        
        # Correct face indices for solid bar
        i_faces = [0, 0, 4, 4, 0, 0, 2, 2, 0, 0, 1, 1]
        j_faces = [1, 2, 6, 7, 5, 4, 7, 6, 3, 7, 5, 6]
        k_faces = [2, 3, 5, 6, 1, 5, 3, 7, 7, 4, 6, 2]
        
        fig.add_trace(
            go.Mesh3d(
                x=vertices_x, y=vertices_y, z=vertices_z,
                i=i_faces, j=j_faces, k=k_faces,
                color=color,
                opacity=0.95,
                name=cls,
                showlegend=False,
                hovertemplate=f'<b>Class {cls}</b><br>Images: {count:,}<extra></extra>',
                flatshading=True,
                lighting=dict(ambient=0.7, diffuse=0.8, specular=0.2),
                lightposition=dict(x=100, y=200, z=300)
            ),
            row=1, col=1
        )
    
    # Update 3D bar scene
    fig.update_scenes(
        dict(
            xaxis=dict(title='Class', tickvals=list(range(n_classes)), ticktext=classes, tickangle=45),
            yaxis=dict(title='', showticklabels=False, range=[-1, 1]),
            zaxis=dict(title='Count'),
            camera=dict(eye=dict(x=1.8, y=1.8, z=0.8))
        ),
        row=1, col=1
    )
    
    # ===== 3D Surface Heatmap =====
    # Create a grid for surface visualization
    grid_size = int(np.ceil(np.sqrt(n_classes)))
    z_grid = np.zeros((grid_size, grid_size))
    labels_grid = [[''] * grid_size for _ in range(grid_size)]
    
    for i, (cls, count) in enumerate(zip(classes, counts)):
        row_idx = i // grid_size
        col_idx = i % grid_size
        if row_idx < grid_size and col_idx < grid_size:
            z_grid[row_idx, col_idx] = count
            labels_grid[row_idx][col_idx] = cls
    
    # Create hover text
    hover_text = [[f'Class: {labels_grid[i][j]}<br>Count: {int(z_grid[i,j]):,}' 
                   if labels_grid[i][j] else '' 
                   for j in range(grid_size)] for i in range(grid_size)]
    
    fig.add_trace(
        go.Surface(
            z=z_grid,
            colorscale='Viridis',
            showscale=True,
            colorbar=dict(title='Image Count', x=1.0),
            hovertemplate='%{text}<extra></extra>',
            text=hover_text,
            contours=dict(
                z=dict(show=True, usecolormap=True, highlightcolor="white", project_z=True)
            )
        ),
        row=1, col=2
    )
    
    # Update surface scene
    fig.update_scenes(
        dict(
            xaxis=dict(title='Grid X'),
            yaxis=dict(title='Grid Y'),
            zaxis=dict(title='Image Count'),
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.2))
        ),
        row=1, col=2
    )
    
    fig.update_layout(
        title=dict(
            text='🔤 ASL Dataset - 3D Class Distribution (Rotate to explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        template='plotly_white',
        height=600,
        margin=dict(l=0, r=0, t=60, b=0)
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], 'class_distribution_3d.png')
    save_and_show_plotly(fig, save_path, '3D Class Distribution')

def visualize_dataset_samples(class_names):
    """Visualize sample images from each class with interactive Plotly grid"""
    
    logger.log("\n📊 Creating interactive dataset sample visualization...")
    
    n_classes = len(class_names)
    n_cols = min(6, n_classes)
    n_rows = (n_classes + n_cols - 1) // n_cols
    
    # Create Plotly subplots
    fig = make_subplots(
        rows=n_rows, cols=n_cols,
        subplot_titles=[f'Class: {c}' for c in class_names],
        horizontal_spacing=0.02,
        vertical_spacing=0.08
    )
    
    for idx, class_name in enumerate(class_names):
        row = idx // n_cols + 1
        col = idx % n_cols + 1
        
        class_dir = os.path.join(CONFIG['train_path'], class_name)
        if os.path.exists(class_dir):
            images = [f for f in os.listdir(class_dir) 
                     if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
            if images:
                img_path = os.path.join(class_dir, images[0])
                img = cv2.imread(img_path)
                if img is not None:
                    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    
                    fig.add_trace(
                        go.Image(z=img_rgb, hovertemplate=f'<b>Class: {class_name}</b><extra></extra>'),
                        row=row, col=col
                    )
    
    fig.update_layout(
        title=dict(
            text='🖼️ Sample Images from Each ASL Class (Click to zoom!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        height=200 * n_rows + 100,
        showlegend=False,
        template='plotly_white'
    )
    
    # Hide axes
    fig.update_xaxes(showticklabels=False, showgrid=False)
    fig.update_yaxes(showticklabels=False, showgrid=False)
    
    save_path = os.path.join(CONFIG['metrics_path'], 'dataset_samples.png')
    save_and_show_plotly(fig, save_path, 'Dataset Samples')

# ==================== MEDIAPIPE HAND CROPPING ====================
class HandCropper:
    """MediaPipe hand detection with smart cropping
    
    Note: MediaPipe uses CPU by default. We minimize its impact by:
    1. Caching cropped images
    2. Using batch processing
    3. Running during data prep, not training
    """
    
    def __init__(self):
        self.mp_hands = mp.solutions.hands
        self.hands = self.mp_hands.Hands(
            static_image_mode=True,
            max_num_hands=1,
            min_detection_confidence=CONFIG['mediapipe_confidence'],
            model_complexity=CONFIG['mediapipe_model_complexity']
        )
        self.margin = CONFIG['mediapipe_margin']
        self.stats = {'total': 0, 'success': 0, 'fallback': 0}
    
    def crop_hand(self, image):
        """Detect and crop hand region"""
        h, w = image.shape[:2]
        self.stats['total'] += 1
        
        # Add padding for edge detection
        pad = 20
        padded = cv2.copyMakeBorder(image, pad, pad, pad, pad, 
                                    cv2.BORDER_CONSTANT, value=[0, 0, 0])
        
        # Convert BGR to RGB for MediaPipe
        rgb = cv2.cvtColor(padded, cv2.COLOR_BGR2RGB)
        results = self.hands.process(rgb)
        
        if results.multi_hand_landmarks:
            landmarks = results.multi_hand_landmarks[0]
            h_pad, w_pad = padded.shape[:2]
            
            x_coords = [int(lm.x * w_pad) for lm in landmarks.landmark]
            y_coords = [int(lm.y * h_pad) for lm in landmarks.landmark]
            
            x_min = max(0, min(x_coords) - self.margin)
            x_max = min(w_pad, max(x_coords) + self.margin)
            y_min = max(0, min(y_coords) - self.margin)
            y_max = min(h_pad, max(y_coords) + self.margin)
            
            if x_max > x_min and y_max > y_min:
                cropped = padded[y_min:y_max, x_min:x_max]
                if cropped.shape[0] > 50 and cropped.shape[1] > 50:
                    self.stats['success'] += 1
                    return cropped, True
        
        # Fallback: center crop
        self.stats['fallback'] += 1
        margin_h, margin_w = int(h * 0.1), int(w * 0.1)
        return image[margin_h:h-margin_h, margin_w:w-margin_w], False
    
    def get_stats(self):
        """Return detection statistics"""
        total = self.stats['total']
        if total == 0:
            return self.stats
        
        self.stats['success_rate'] = self.stats['success'] / total * 100
        self.stats['fallback_rate'] = self.stats['fallback'] / total * 100
        return self.stats
    
    def close(self):
        """Clean up MediaPipe resources"""
        if hasattr(self, 'hands') and self.hands:
            self.hands.close()

def process_dataset_with_mediapipe():
    """Process and cache dataset with MediaPipe hand cropping
    
    NOTE: MediaPipe uses TensorFlow Lite which runs on CPU only.
    This is expected behavior - GPU is used for model training, not MediaPipe.
    We optimize by caching results so this only runs once.
    """
    
    logger.section("MEDIAPIPE HAND CROPPING")
    
    logger.log("⚠️ NOTE: MediaPipe uses TFLite (CPU-only). This is normal.")
    logger.log("⚠️ GPU will be used for model training after preprocessing.")
    logger.log("")
    
    # Check if cached crops exist
    if CONFIG['use_cached_crops']:
        train_exists = os.path.exists(CONFIG['cropped_train_path'])
        has_classes = False
        if train_exists:
            subdirs = [d for d in os.listdir(CONFIG['cropped_train_path']) 
                      if os.path.isdir(os.path.join(CONFIG['cropped_train_path'], d))]
            has_classes = len(subdirs) > 0
        
        if has_classes:
            class_names = sorted([d for d in subdirs if d not in CONFIG['skip_folders']])
            logger.log(f"✅ Using cached crops from: {CONFIG['cropped_train_path']}")
            logger.log(f"✅ Found {len(class_names)} classes")
            
            # Count cached images
            total_cached = 0
            for cls in class_names:
                cls_dir = os.path.join(CONFIG['cropped_train_path'], cls)
                if os.path.exists(cls_dir):
                    total_cached += len([f for f in os.listdir(cls_dir) if f.endswith(('.jpg', '.png'))])
            logger.log(f"✅ Total cached images: {total_cached:,}")
            return class_names
    
    # Check source exists
    if not os.path.exists(CONFIG['train_path']):
        logger.log(f"❌ Training path not found: {CONFIG['train_path']}")
        return []
    
    cropper = HandCropper()
    
    # Get valid classes
    all_folders = [d for d in os.listdir(CONFIG['train_path']) 
                   if os.path.isdir(os.path.join(CONFIG['train_path'], d))]
    class_names = sorted([c for c in all_folders if c not in CONFIG['skip_folders']])
    
    logger.log(f"\n📊 Processing {len(class_names)} classes...")
    logger.log(f"📊 Images per class: {CONFIG['images_per_class'] or 'ALL'}")
    
    # Calculate total work
    total_images_to_process = 0
    for class_name in class_names:
        src_dir = os.path.join(CONFIG['train_path'], class_name)
        all_images = [f for f in os.listdir(src_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        if CONFIG['images_per_class']:
            total_images_to_process += min(len(all_images), CONFIG['images_per_class'])
        else:
            total_images_to_process += len(all_images)
    
    logger.log(f"📊 Total images to process: {total_images_to_process:,}")
    logger.log(f"📊 Estimated time: ~{total_images_to_process * 0.03 / 60:.1f} minutes")
    
    # Process training data with detailed progress
    logger.log("\n🔄 Processing TRAINING data...")
    
    processing_stats = {
        'total_processed': 0,
        'successful': 0,
        'fallback': 0,
        'failed': 0,
        'per_class': {}
    }
    
    start_time = time.time()
    
    for class_idx, class_name in enumerate(class_names):
        class_start = time.time()
        src_dir = os.path.join(CONFIG['train_path'], class_name)
        dst_dir = os.path.join(CONFIG['cropped_train_path'], class_name)
        os.makedirs(dst_dir, exist_ok=True)
        
        all_images = [f for f in os.listdir(src_dir) 
                     if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        
        # Sample images if configured
        if CONFIG['images_per_class'] and len(all_images) > CONFIG['images_per_class']:
            np.random.seed(42)
            images = list(np.random.choice(all_images, CONFIG['images_per_class'], replace=False))
        else:
            images = all_images
        
        saved_count = 0
        for img_name in images:
            img_path = os.path.join(src_dir, img_name)
            img = cv2.imread(img_path)
            if img is None:
                continue
            
            cropped, success = cropper.crop_hand(img)
            
            # Skip fallback crops if required
            if CONFIG['require_mediapipe_detection'] and not success:
                continue
            
            try:
                cropped_resized = cv2.resize(cropped, CONFIG['img_size'])
                save_path = os.path.join(dst_dir, img_name)
                cv2.imwrite(save_path, cropped_resized)
                saved_count += 1
                if success:
                    processing_stats['successful'] += 1
                else:
                    processing_stats['fallback'] += 1
            except Exception as e:
                processing_stats['failed'] += 1
                continue
        
        processing_stats['total_processed'] += len(images)
        processing_stats['per_class'][class_name] = saved_count
        
        # Print per-class progress
        class_time = time.time() - class_start
        elapsed = time.time() - start_time
        remaining_classes = len(class_names) - (class_idx + 1)
        eta = (elapsed / (class_idx + 1)) * remaining_classes if class_idx > 0 else 0
        
        print(f"  [{class_idx+1:2d}/{len(class_names)}] {class_name}: {saved_count:4d} saved | "
              f"Time: {class_time:.1f}s | ETA: {eta/60:.1f}min", flush=True)
        
        # Also save some to test set
        if CONFIG['test_images_per_class_from_train'] > 0:
            remaining = [f for f in all_images if f not in images[:saved_count]]
            if remaining:
                n_take = min(CONFIG['test_images_per_class_from_train'], len(remaining))
                np.random.seed(42)
                test_images = list(np.random.choice(remaining, n_take, replace=False))
                
                dst_test_dir = os.path.join(CONFIG['cropped_test_path'], class_name)
                os.makedirs(dst_test_dir, exist_ok=True)
                
                for img_name in test_images:
                    img_path = os.path.join(src_dir, img_name)
                    img = cv2.imread(img_path)
                    if img is None:
                        continue
                    
                    cropped, success = cropper.crop_hand(img)
                    if CONFIG['require_mediapipe_detection'] and not success:
                        continue
                    
                    try:
                        cropped_resized = cv2.resize(cropped, CONFIG['img_size'])
                        save_path = os.path.join(dst_test_dir, img_name)
                        cv2.imwrite(save_path, cropped_resized)
                    except:
                        continue
    
    # Print processing summary
    total_time = time.time() - start_time
    logger.log(f"\n{'='*60}")
    logger.log(f"📊 MEDIAPIPE PROCESSING COMPLETE")
    logger.log(f"{'='*60}")
    logger.log(f"   Total time: {total_time/60:.1f} minutes")
    logger.log(f"   Images processed: {processing_stats['total_processed']:,}")
    logger.log(f"   Successful detections: {processing_stats['successful']:,}")
    logger.log(f"   Fallback crops: {processing_stats['fallback']:,}")
    logger.log(f"   Failed: {processing_stats['failed']:,}")
    logger.log(f"   Processing speed: {processing_stats['total_processed']/total_time:.1f} img/sec")
    
    # Per-class summary
    logger.log(f"\n📊 Per-Class Results:")
    logger.log(f"{'-'*40}")
    for cls, count in processing_stats['per_class'].items():
        logger.log(f"   {cls}: {count:,} images saved")
    logger.log(f"{'-'*40}")
    
    # Save processing stats
    stats_path = os.path.join(CONFIG['metrics_path'], 'mediapipe_processing_stats.json')
    with open(stats_path, 'w') as f:
        json.dump(processing_stats, f, indent=2)
    logger.log(f"💾 Processing stats saved to: {stats_path}")
    
    # Process test data from test folder
    if os.path.exists(CONFIG['test_path']):
        logger.log("\n🔄 Processing TEST data...")
        test_files = [f for f in os.listdir(CONFIG['test_path']) 
                     if f.lower().endswith('.jpg') and '_test' in f]
        
        test_saved = 0
        for test_file in tqdm(test_files, desc="Test images"):
            if test_file in CONFIG['skip_test_files']:
                continue
            
            class_name = test_file.split('_test')[0]
            if class_name not in class_names:
                continue
            
            src_path = os.path.join(CONFIG['test_path'], test_file)
            img = cv2.imread(src_path)
            if img is None:
                continue
            
            cropped, success = cropper.crop_hand(img)
            if CONFIG['require_mediapipe_detection'] and not success:
                continue
            
            try:
                cropped_resized = cv2.resize(cropped, CONFIG['img_size'])
                dst_dir = os.path.join(CONFIG['cropped_test_path'], class_name)
                os.makedirs(dst_dir, exist_ok=True)
                save_path = os.path.join(dst_dir, test_file)
                cv2.imwrite(save_path, cropped_resized)
                test_saved += 1
            except:
                continue
        
        logger.log(f"   Test images saved: {test_saved}")
    
    # Print statistics
    stats = cropper.get_stats()
    logger.log(f"\n📊 MediaPipe Detection Statistics:")
    logger.log(f"   Total processed: {stats['total']:,}")
    logger.log(f"   Successful detections: {stats['success']:,} ({stats.get('success_rate', 0):.1f}%)")
    logger.log(f"   Fallback crops: {stats['fallback']:,} ({stats.get('fallback_rate', 0):.1f}%)")
    
    cropper.close()
    
    # Save sample visualization
    visualize_cropped_samples(class_names)
    
    # Visualize processing statistics
    visualize_mediapipe_stats(processing_stats, class_names)
    
    return class_names

def visualize_mediapipe_stats(stats, class_names):
    """Visualize MediaPipe processing statistics with interactive Plotly"""
    
    logger.log("\n📊 Creating interactive MediaPipe processing visualization...")
    
    # Create subplots
    fig = make_subplots(
        rows=1, cols=2,
        specs=[[{"type": "bar"}, {"type": "pie"}]],
        subplot_titles=['Images Saved per Class', 'MediaPipe Detection Results']
    )
    
    # Plot 1: Images saved per class (3D-style bar)
    classes = list(stats['per_class'].keys())
    counts = list(stats['per_class'].values())
    
    fig.add_trace(
        go.Bar(
            x=classes,
            y=counts,
            marker=dict(
                color=counts,
                colorscale='Viridis',
                showscale=True,
                colorbar=dict(title='Count', x=0.45)
            ),
            text=counts,
            textposition='outside',
            hovertemplate='<b>Class %{x}</b><br>Saved: %{y:,}<extra></extra>',
            name='Images Saved'
        ),
        row=1, col=1
    )
    
    # Plot 2: Processing breakdown pie chart
    labels = ['✅ Successful Detection', '⚠️ Fallback Crop', '❌ Failed']
    values = [stats['successful'], stats['fallback'], stats['failed']]
    colors_pie = ['#2ecc71', '#f39c12', '#e74c3c']
    
    # Filter out zeros
    filtered = [(l, v, c) for l, v, c in zip(labels, values, colors_pie) if v > 0]
    if filtered:
        labels, values, colors_pie = zip(*filtered)
        fig.add_trace(
            go.Pie(
                labels=labels,
                values=values,
                marker=dict(colors=colors_pie),
                textinfo='percent+label',
                hovertemplate='<b>%{label}</b><br>Count: %{value:,}<br>%{percent}<extra></extra>',
                hole=0.3  # Donut chart for modern look
            ),
            row=1, col=2
        )
    
    fig.update_layout(
        title=dict(
            text='🤖 MediaPipe Processing Statistics (Interactive)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        height=500,
        template='plotly_white',
        showlegend=False
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], 'mediapipe_processing_stats.png')
    save_and_show_plotly(fig, save_path, 'MediaPipe Processing Stats')

def visualize_cropped_samples(class_names):
    """Visualize sample cropped images with interactive Plotly"""
    
    logger.log("\n📊 Creating interactive cropped samples visualization...")
    
    sample_classes = class_names[:6]
    n_cols = min(6, len(sample_classes))
    
    # Create subplots for 2 rows x n_cols
    fig = make_subplots(
        rows=2, cols=n_cols,
        subplot_titles=[f'{c}' for c in sample_classes] + [''] * n_cols,
        horizontal_spacing=0.02,
        vertical_spacing=0.08
    )
    
    for idx, class_name in enumerate(sample_classes):
        class_dir = os.path.join(CONFIG['cropped_train_path'], class_name)
        if os.path.exists(class_dir):
            images = os.listdir(class_dir)
            if images:
                for row in range(2):
                    if row < len(images):
                        img_path = os.path.join(class_dir, images[row])
                        img = cv2.imread(img_path)
                        if img is not None:
                            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                            fig.add_trace(
                                go.Image(z=img_rgb, hovertemplate=f'<b>Class: {class_name}</b><extra></extra>'),
                                row=row+1, col=idx+1
                            )
    
    fig.update_layout(
        title=dict(
            text='✂️ Sample Cropped Training Images (Click to zoom!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        height=500,
        showlegend=False,
        template='plotly_white'
    )
    
    fig.update_xaxes(showticklabels=False, showgrid=False)
    fig.update_yaxes(showticklabels=False, showgrid=False)
    
    save_path = os.path.join(CONFIG['metrics_path'], 'sample_cropped_images.png')
    save_and_show_plotly(fig, save_path, 'Sample Cropped Images')

# ==================== DATA LOADING ====================
class AugmentedDataGenerator(keras.utils.Sequence):
    """Custom data generator with model-specific preprocessing"""
    
    def __init__(self, image_paths, labels, batch_size, model_name,
                 augmentation=None, shuffle=True):
        self.image_paths = image_paths
        self.labels = labels
        self.batch_size = batch_size
        self.model_name = model_name
        self.augmentation = augmentation
        self.shuffle = shuffle
        self.indices = np.arange(len(self.image_paths))
        
        # Set preprocessing function based on model
        if model_name == 'EfficientNetV2B3':
            self.preprocess_fn = effnet_preprocess
        elif model_name == 'ResNet50':
            self.preprocess_fn = resnet_preprocess
        elif model_name == 'InceptionV3':
            self.preprocess_fn = inception_preprocess
        else:
            self.preprocess_fn = None
        
        self.on_epoch_end()
    
    def __len__(self):
        return int(np.ceil(len(self.image_paths) / self.batch_size))
    
    def __getitem__(self, idx):
        batch_indices = self.indices[idx * self.batch_size:(idx + 1) * self.batch_size]
        batch_paths = [self.image_paths[i] for i in batch_indices]
        batch_labels = [self.labels[i] for i in batch_indices]
        
        X, y = [], []
        for path, label in zip(batch_paths, batch_labels):
            img = cv2.imread(path)
            if img is None:
                img = np.zeros((CONFIG['img_size'][0], CONFIG['img_size'][1], 3), dtype=np.uint8)
            else:
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            
            # Apply augmentation
            if self.augmentation:
                try:
                    img = self.augmentation(image=img)['image']
                except:
                    pass
            
            # Ensure correct size
            img = cv2.resize(img, CONFIG['img_size'])
            
            # Apply model-specific preprocessing
            if self.preprocess_fn:
                img = self.preprocess_fn(img.astype(np.float32))
            else:
                img = img.astype(np.float32) / 255.0
            
            X.append(img)
            y.append(label)
        
        return np.array(X), np.array(y)
    
    def on_epoch_end(self):
        if self.shuffle:
            np.random.shuffle(self.indices)

def load_dataset(cropped_path, class_names):
    """Load cropped dataset"""
    
    class_to_idx = {name: idx for idx, name in enumerate(class_names)}
    image_paths, labels = [], []
    
    for class_name in class_names:
        class_dir = os.path.join(cropped_path, class_name)
        if not os.path.exists(class_dir):
            continue
        
        images = [f for f in os.listdir(class_dir) 
                 if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        
        for img_name in images:
            image_paths.append(os.path.join(class_dir, img_name))
            labels.append(class_to_idx[class_name])
    
    return np.array(image_paths), np.array(labels)

# ==================== AUGMENTATION PIPELINES ====================
def get_augmentation_pipeline(model_name):
    """Model-specific augmentation pipelines"""
    
    if model_name == 'ResNet50':
        return A.Compose([
            A.Rotate(limit=25, p=0.8),
            A.Affine(scale=(0.85, 1.15), translate_percent=(-0.15, 0.15),
                    rotate=(-20, 20), shear=(-10, 10), p=0.7),
            A.HorizontalFlip(p=0.5),
            A.OneOf([
                A.MotionBlur(blur_limit=5, p=1.0),
                A.GaussianBlur(blur_limit=5, p=1.0),
            ], p=0.4),
            A.RandomBrightnessContrast(brightness_limit=0.25, contrast_limit=0.25, p=0.5),
            A.CLAHE(clip_limit=2.0, p=0.3),
            A.GaussNoise(var_limit=(10.0, 40.0), p=0.25),
            A.CoarseDropout(num_holes_range=(1, 3), hole_height_range=(8, 16),
                           hole_width_range=(8, 16), p=0.25),
        ])
    
    elif model_name == 'EfficientNetV2B3':
        return A.Compose([
            A.Rotate(limit=30, p=0.9),
            A.Affine(scale=(0.80, 1.2), translate_percent=(-0.12, 0.12),
                    rotate=(-18, 18), p=0.7),
            A.HorizontalFlip(p=0.5),
            A.OneOf([
                A.GaussianBlur(blur_limit=3, p=1.0),
                A.MotionBlur(blur_limit=3, p=1.0),
            ], p=0.35),
            A.RandomBrightnessContrast(brightness_limit=0.25, contrast_limit=0.25, p=0.5),
            A.CLAHE(clip_limit=2.0, p=0.3),
            A.GaussNoise(var_limit=(5.0, 30.0), p=0.2),
            A.CoarseDropout(num_holes_range=(1, 4), hole_height_range=(6, 20),
                           hole_width_range=(6, 20), p=0.25),
        ])
    
    elif model_name == 'InceptionV3':
        return A.Compose([
            A.Rotate(limit=15, p=0.6),
            A.Affine(scale=(0.9, 1.1), translate_percent=(-0.1, 0.1),
                    rotate=(-10, 10), p=0.5),
            A.RandomBrightnessContrast(brightness_limit=0.15, contrast_limit=0.15, p=0.35),
            A.CLAHE(clip_limit=1.5, p=0.2),
        ])
    
    return None

# ==================== MODEL BUILDERS ====================
def build_resnet50(num_classes):
    """ResNet50 with fine-tuning"""
    base = ResNet50(weights='imagenet', include_top=False, 
                    input_shape=(*CONFIG['img_size'], 3))
    
    for layer in base.layers[:-30]:
        layer.trainable = False
    
    model = models.Sequential([
        base,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(512, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.4),
        layers.Dense(256, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax')
    ], name='ResNet50_ASL')
    
    return model

def build_efficientnet(num_classes):
    """EfficientNetV2B3 with fine-tuning"""
    try:
        base = EfficientNetV2B3(weights='imagenet', include_top=False, 
                                input_shape=(*CONFIG['img_size'], 3))
    except:
        base = EfficientNetV2B3(weights=None, include_top=False, 
                                input_shape=(*CONFIG['img_size'], 3))
    
    for layer in base.layers[:-80]:
        layer.trainable = False
    
    model = models.Sequential([
        base,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(512, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.35),
        layers.Dense(256, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
        layers.Dropout(0.25),
        layers.Dense(num_classes, activation='softmax')
    ], name='EfficientNetV2B3_ASL')
    
    return model

def build_inception(num_classes):
    """InceptionV3 with fine-tuning"""
    base = InceptionV3(weights='imagenet', include_top=False, 
                       input_shape=(*CONFIG['img_size'], 3))
    
    for layer in base.layers[:-60]:
        layer.trainable = False
    
    model = models.Sequential([
        base,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dropout(0.45),
        layers.Dense(512, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.35),
        layers.Dense(256, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
        layers.Dropout(0.25),
        layers.Dense(num_classes, activation='softmax')
    ], name='InceptionV3_ASL')
    
    return model

MODEL_BUILDERS = {
    'ResNet50': build_resnet50,
    'EfficientNetV2B3': build_efficientnet,
    'InceptionV3': build_inception
}

# ==================== MODEL INFO COLLECTION ====================
def collect_model_info(num_classes):
    """Collect comprehensive information about all models BEFORE training
    
    This function:
    1. Creates each model architecture
    2. Collects detailed layer info, parameter counts
    3. Tests if existing models can be loaded
    4. Stores all info in a structured format
    5. Saves visualizations of model architectures
    """
    
    logger.section("MODEL INFORMATION COLLECTION")
    
    model_info = {}
    
    for model_name in CONFIG['train_models']:
        logger.log(f"\n{'='*60}")
        logger.log(f"📊 ANALYZING: {model_name}")
        logger.log(f"{'='*60}")
        
        info = {
            'name': model_name,
            'status': 'unknown',
            'existing_model_path': None,
            'existing_model_found': False,
            'will_resume': False,
            'total_params': 0,
            'trainable_params': 0,
            'non_trainable_params': 0,
            'trainable_ratio': 0,
            'num_layers': 0,
            'base_model_layers': 0,
            'custom_layers': 0,
            'input_shape': None,
            'output_shape': None,
            'layer_summary': [],
            'memory_estimate_mb': 0
        }
        
        # Check for existing model
        existing_path = CONFIG['existing_models'].get(model_name)
        info['existing_model_path'] = existing_path
        
        if existing_path and os.path.exists(existing_path):
            info['existing_model_found'] = True
            info['will_resume'] = True
            info['status'] = 'RESUME - Existing model found'
            logger.log(f"✅ EXISTING MODEL FOUND: {existing_path}")
            
            # Try to load and get info from existing model using new loader
            try:
                existing_model, format_type = load_model_any_format(existing_path, model_name)
                
                if existing_model is not None:
                    info['total_params'] = existing_model.count_params()
                    info['trainable_params'] = sum([tf.size(w).numpy() for w in existing_model.trainable_weights])
                    info['non_trainable_params'] = info['total_params'] - info['trainable_params']
                    info['num_layers'] = len(existing_model.layers)
                    info['input_shape'] = str(existing_model.input_shape)
                    info['output_shape'] = str(existing_model.output_shape)
                    info['model_format'] = format_type
                    
                    # If it's SavedModel (inference-only), mark for scratch training
                    if format_type == 'savedmodel':
                        info['will_resume'] = False
                        info['status'] = 'SCRATCH - SavedModel (inference-only)'
                        logger.log(f"   ⚠️ SavedModel format is inference-only, will train from scratch")
                    
                    del existing_model
                    keras.backend.clear_session()
                else:
                    # format_type contains error message
                    raise Exception(format_type)
            except Exception as e:
                logger.log(f"⚠️ Could not load existing model: {e}")
                info['will_resume'] = False
                info['status'] = 'SCRATCH - Load failed'
        else:
            info['status'] = 'SCRATCH - No existing model'
            logger.log(f"🆕 NO EXISTING MODEL at: {existing_path}")
            logger.log(f"   Will train from SCRATCH")
        
        # Build fresh model to get architecture info
        try:
            model = MODEL_BUILDERS[model_name](num_classes)
            
            info['total_params'] = model.count_params()
            info['trainable_params'] = sum([tf.size(w).numpy() for w in model.trainable_weights])
            info['non_trainable_params'] = info['total_params'] - info['trainable_params']
            info['trainable_ratio'] = info['trainable_params'] / info['total_params'] * 100 if info['total_params'] > 0 else 0
            info['num_layers'] = len(model.layers)
            info['input_shape'] = str(model.input_shape)
            info['output_shape'] = str(model.output_shape)
            
            # Estimate memory (rough: 4 bytes per parameter for float32)
            info['memory_estimate_mb'] = (info['total_params'] * 4) / (1024 * 1024)
            
            # Count base vs custom layers
            for layer in model.layers:
                if hasattr(layer, 'layers'):  # It's a nested model (base)
                    info['base_model_layers'] = len(layer.layers)
                else:
                    info['custom_layers'] += 1
            
            # Collect layer summary (last 15 layers)
            for layer in model.layers[-15:]:
                # Safely get output shape (TF 2.18+ compatibility)
                try:
                    out_shape = str(layer.output.shape) if hasattr(layer, 'output') else 'N/A'
                except:
                    out_shape = 'N/A'
                
                layer_info = {
                    'name': layer.name,
                    'type': layer.__class__.__name__,
                    'output_shape': out_shape,
                    'trainable': layer.trainable,
                    'params': layer.count_params()
                }
                info['layer_summary'].append(layer_info)
            
            # Print detailed info
            logger.log(f"\n📊 Architecture Details:")
            logger.log(f"   Input Shape:          {info['input_shape']}")
            logger.log(f"   Output Shape:         {info['output_shape']}")
            logger.log(f"   Total Parameters:     {info['total_params']:,}")
            logger.log(f"   Trainable Params:     {info['trainable_params']:,}")
            logger.log(f"   Non-trainable Params: {info['non_trainable_params']:,}")
            logger.log(f"   Trainable Ratio:      {info['trainable_ratio']:.1f}%")
            logger.log(f"   Total Layers:         {info['num_layers']}")
            logger.log(f"   Base Model Layers:    {info['base_model_layers']}")
            logger.log(f"   Custom Layers:        {info['custom_layers']}")
            logger.log(f"   Est. Memory:          {info['memory_estimate_mb']:.1f} MB")
            
            logger.log(f"\n📋 Layer Summary (last 15):")
            logger.log("-" * 70)
            for li in info['layer_summary']:
                trainable_mark = "✓" if li['trainable'] else "✗"
                logger.log(f"   [{trainable_mark}] {li['name']:<30} {li['type']:<20} {li['params']:>10,} params")
            logger.log("-" * 70)
            
            # Cleanup
            del model
            keras.backend.clear_session()
            gc.collect()
            
        except Exception as e:
            logger.log(f"❌ Error building {model_name}: {e}")
            info['status'] = f'ERROR - {str(e)}'
        
        model_info[model_name] = info
    
    # Save model info to JSON
    model_info_path = os.path.join(CONFIG['metrics_path'], 'model_info.json')
    
    # Convert to JSON-serializable format
    save_info = {}
    for name, info in model_info.items():
        save_info[name] = {k: v for k, v in info.items()}
    
    with open(model_info_path, 'w') as f:
        json.dump(save_info, f, indent=2, default=str)
    
    logger.log(f"\n💾 Model info saved to: {model_info_path}")
    
    # Create comparison visualization
    visualize_model_architecture_comparison(model_info)
    
    # Print summary table
    logger.log(f"\n{'='*80}")
    logger.log(f"📊 MODEL SUMMARY TABLE")
    logger.log(f"{'='*80}")
    logger.log(f"{'Model':<20} {'Status':<25} {'Params':>15} {'Trainable':>15}")
    logger.log(f"{'-'*80}")
    for name, info in model_info.items():
        logger.log(f"{name:<20} {info['status']:<25} {info['total_params']:>15,} {info['trainable_params']:>15,}")
    logger.log(f"{'='*80}")
    
    return model_info

def visualize_model_architecture_comparison(model_info):
    """Create 3D INTERACTIVE visualization comparing model parameters with SOLID bars"""
    
    logger.log("\n📊 Creating 3D model parameters comparison...")
    
    models = list(model_info.keys())
    n_models = len(models)
    
    # Data preparation
    total_params = [model_info[m]['total_params'] / 1e6 for m in models]
    trainable_params = [model_info[m]['trainable_params'] / 1e6 for m in models]
    non_trainable = [t - tr for t, tr in zip(total_params, trainable_params)]
    trainable_ratio = [model_info[m]['trainable_ratio'] for m in models]
    
    # Create 3D figure with SOLID bars for each model
    fig = go.Figure()
    
    max_params = max(total_params)
    
    for i, model in enumerate(models):
        # Trainable params bar (green) - left side
        bar_train = create_3d_bar_mesh(
            x_center=i - 0.2,
            y_center=0,
            z_height=trainable_params[i],
            bar_width=0.35,
            bar_depth=0.6,
            color='#2ecc71',
            opacity=0.95,
            name='Trainable' if i == 0 else '',
            hovertemplate=f'<b>{model}</b><br>Trainable: {trainable_params[i]:.1f}M<extra></extra>'
        )
        bar_train.update(showlegend=(i == 0), legendgroup='Trainable')
        fig.add_trace(bar_train)
        
        # Frozen params bar (blue) - right side
        bar_frozen = create_3d_bar_mesh(
            x_center=i + 0.2,
            y_center=0,
            z_height=non_trainable[i],
            bar_width=0.35,
            bar_depth=0.6,
            color='#3498db',
            opacity=0.95,
            name='Frozen' if i == 0 else '',
            hovertemplate=f'<b>{model}</b><br>Frozen: {non_trainable[i]:.1f}M<extra></extra>'
        )
        bar_frozen.update(showlegend=(i == 0), legendgroup='Frozen')
        fig.add_trace(bar_frozen)
    
    # Add spheres at top showing total
    fig.add_trace(go.Scatter3d(
        x=list(range(n_models)),
        y=[0] * n_models,
        z=[t + max_params * 0.08 for t in total_params],
        mode='markers+text',
        marker=dict(size=14, color='#e74c3c', symbol='diamond'),
        text=[f'{t:.1f}M' for t in total_params],
        textposition='top center',
        textfont=dict(size=12, color='black', family='Arial Black'),
        name='Total',
        hovertemplate='<b>%{text}</b><br>Total Params<extra></extra>'
    ))
    
    fig.update_layout(
        title=dict(
            text='🏗️ 3D Model Parameters Comparison (Rotate to explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(title='Model', tickvals=list(range(n_models)), ticktext=models),
            yaxis=dict(title='', showticklabels=False, range=[-1, 1]),
            zaxis=dict(title='Parameters (Millions)'),
            camera=dict(eye=dict(x=1.5, y=1.5, z=0.8)),
            bgcolor='rgba(250,250,250,0.9)'
        ),
        height=650,
        template='plotly_white',
        legend=dict(orientation='h', y=1.05)
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], 'model_architecture_comparison.png')
    save_and_show_plotly(fig, save_path, 'Model Parameters Comparison')

# ==================== BASELINE CONFUSION MATRIX VISUALIZATION ====================
def visualize_baseline_confusion_matrix(cm, class_names, model_name):
    """Create 3D INTERACTIVE confusion matrix visualization for baseline testing"""
    
    logger.log(f"\n📊 Creating 3D baseline confusion matrix for {model_name}...")
    
    n_classes = len(class_names)
    
    # Create 3D Surface plot for confusion matrix
    fig = make_subplots(
        rows=1, cols=2,
        specs=[[{"type": "surface"}, {"type": "surface"}]],
        subplot_titles=[f'Confusion Matrix (Raw Counts)', f'Normalized Confusion Matrix'],
        horizontal_spacing=0.05
    )
    
    # 3D Surface for raw confusion matrix
    fig.add_trace(
        go.Surface(
            z=cm,
            x=list(range(n_classes)),
            y=list(range(n_classes)),
            colorscale='Blues',
            showscale=True,
            colorbar=dict(title='Count', x=0.45, len=0.8),
            hovertemplate='True: %{y}<br>Pred: %{x}<br>Count: %{z}<extra></extra>',
            contours=dict(
                z=dict(show=True, usecolormap=True, highlightcolor="white", project_z=True)
            )
        ),
        row=1, col=1
    )
    
    # Create normalized confusion matrix
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    cm_normalized = np.nan_to_num(cm_normalized)
    
    # 3D Surface for normalized confusion matrix
    fig.add_trace(
        go.Surface(
            z=cm_normalized * 100,  # Convert to percentage
            x=list(range(n_classes)),
            y=list(range(n_classes)),
            colorscale='RdYlGn',
            showscale=True,
            colorbar=dict(title='%', x=1.0, len=0.8),
            hovertemplate='True: %{y}<br>Pred: %{x}<br>Rate: %{z:.1f}%<extra></extra>',
            contours=dict(
                z=dict(show=True, usecolormap=True, highlightcolor="white", project_z=True)
            )
        ),
        row=1, col=2
    )
    
    # Update both scenes
    for col in [1, 2]:
        fig.update_scenes(
            dict(
                xaxis=dict(title='Predicted', tickvals=list(range(n_classes)), ticktext=class_names, tickangle=45),
                yaxis=dict(title='True', tickvals=list(range(n_classes)), ticktext=class_names, tickangle=-45),
                zaxis=dict(title='Count' if col == 1 else 'Rate %'),
                camera=dict(eye=dict(x=1.5, y=-1.5, z=1.0))
            ),
            row=1, col=col
        )
    
    fig.update_layout(
        title=dict(
            text=f'🎯 {model_name} - 3D Confusion Matrix (ALL {n_classes} Classes) - Rotate to explore!',
            font=dict(size=18, color='#333'),
            x=0.5
        ),
        height=700,
        template='plotly_white'
    )
    
    cm_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_baseline_confusion_matrix.png')
    save_and_show_plotly(fig, cm_path, f'{model_name} 3D Baseline Confusion Matrix')
    
    # Also create a 3D bar chart version showing diagonal (correct predictions) with SOLID bars
    fig_diag = go.Figure()
    
    diagonal = np.diag(cm)
    max_val = diagonal.max() if diagonal.max() > 0 else 1
    
    for i, (cls, count) in enumerate(zip(class_names, diagonal)):
        norm = count / max_val
        
        # Green for high accuracy, red for low
        r = int(255 * (1 - norm))
        g = int(200 * norm)
        b = 50
        color = f'rgb({r}, {g}, {b})'
        
        # Create solid 3D bar using helper
        bar = create_3d_bar_mesh(
            x_center=i,
            y_center=0,
            z_height=count,
            bar_width=0.7,
            bar_depth=0.7,
            color=color,
            opacity=0.95,
            name=cls,
            hovertemplate=f'<b>Class {cls}</b><br>Correct: {count:,}<extra></extra>'
        )
        fig_diag.add_trace(bar)
    
    fig_diag.update_layout(
        title=dict(
            text=f'🎯 {model_name} - Correct Predictions Per Class (3D)',
            font=dict(size=18, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(title='Class', tickvals=list(range(n_classes)), ticktext=class_names, tickangle=45),
            yaxis=dict(title='', showticklabels=False, range=[-1, 1]),
            zaxis=dict(title='Correct Count'),
            camera=dict(eye=dict(x=1.8, y=1.8, z=0.8)),
            bgcolor='rgba(250,250,250,0.9)'
        ),
        height=650,
        template='plotly_white'
    )
    
    diag_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_baseline_correct_predictions_3d.png')
    save_and_show_plotly(fig_diag, diag_path, f'{model_name} Correct Predictions 3D')

# ==================== PRE-TRAINING BASELINE TEST ====================
def test_models_before_training(test_paths, test_labels, class_names, model_info):
    """Test existing models BEFORE training to get baseline metrics
    
    This helps understand:
    1. Current model performance (if resuming)
    2. What we're starting from
    3. Comparison after training
    
    IMPORTANT: This function checks for class count mismatches and recommends
    training from scratch if the model was trained on different data.
    """
    
    logger.section("PRE-TRAINING BASELINE TEST")
    
    baseline_results = {}
    num_expected_classes = len(class_names)
    
    logger.log(f"\n📊 Expected classes in current dataset: {num_expected_classes}")
    logger.log(f"   Class names: {class_names[:5]}... (showing first 5)")
    
    for model_name in CONFIG['train_models']:
        info = model_info.get(model_name, {})
        
        if not info.get('existing_model_found', False):
            logger.log(f"\n⏭️ {model_name}: No existing model - will train from scratch")
            baseline_results[model_name] = {'baseline_tested': False, 'reason': 'No existing model', 'force_scratch': True}
            continue
        
        existing_path = info.get('existing_model_path')
        
        logger.log(f"\n{'='*60}")
        logger.log(f"🔬 TESTING BASELINE: {model_name}")
        logger.log(f"{'='*60}")
        logger.log(f"   Loading from: {existing_path}")
        
        try:
            # Use the new loader that supports all formats including SavedModel
            model, format_type = load_model_any_format(existing_path, model_name)
            
            if model is None:
                # format_type contains error message
                raise Exception(format_type)
            
            # Check if this is a SavedModel (inference-only)
            is_inference_only = (format_type == 'savedmodel')
            
            # Set preprocessing function
            if model_name == 'EfficientNetV2B3':
                preprocess_fn = effnet_preprocess
            elif model_name == 'ResNet50':
                preprocess_fn = resnet_preprocess
            elif model_name == 'InceptionV3':
                preprocess_fn = inception_preprocess
            else:
                preprocess_fn = None
            
            # Load ALL test data (no limit - we want comprehensive results)
            X_test, y_test = [], []
            logger.log(f"   Loading ALL {len(test_paths)} test samples...")
            
            # DEBUG: Print preprocessing info
            logger.log(f"   Preprocessing function: {preprocess_fn.__name__ if preprocess_fn else 'None (using /255.0)'}")
            
            for path, label in zip(test_paths, test_labels):  # Use ALL samples
                img = cv2.imread(path)
                if img is None:
                    continue
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                img = cv2.resize(img, CONFIG['img_size'])
                
                if preprocess_fn:
                    img = preprocess_fn(img.astype(np.float32))
                else:
                    img = img.astype(np.float32) / 255.0
                
                X_test.append(img)
                y_test.append(label)
            
            X_test = np.array(X_test)
            y_test = np.array(y_test)
            
            if len(X_test) == 0:
                logger.log(f"⚠️ No test samples available")
                baseline_results[model_name] = {'baseline_tested': False, 'reason': 'No test samples'}
                continue
            
            # DEBUG: Print data statistics
            logger.log(f"   Data shape: {X_test.shape}")
            logger.log(f"   Data range: [{X_test.min():.2f}, {X_test.max():.2f}]")
            logger.log(f"   Data mean: {X_test.mean():.2f}")
            logger.log(f"   Classes represented: {len(np.unique(y_test))} of {num_expected_classes}")
            logger.log(f"   Testing on {len(X_test)} samples (ALL available)...")
            
            # Check model output shape matches number of classes
            model_output_classes = model.output_shape[-1]
            
            if model_output_classes != num_expected_classes:
                logger.log(f"")
                logger.log(f"⚠️ " + "="*50)
                logger.log(f"⚠️ CLASS COUNT MISMATCH DETECTED!")
                logger.log(f"⚠️ " + "="*50)
                logger.log(f"⚠️ Model outputs: {model_output_classes} classes")
                logger.log(f"⚠️ Dataset has:   {num_expected_classes} classes")
                logger.log(f"⚠️ ")
                logger.log(f"⚠️ This model was likely trained on a different version")
                logger.log(f"⚠️ of the dataset (before/after MediaPipe filtering).")
                logger.log(f"⚠️ ")
                logger.log(f"⚠️ → Will FORCE training from SCRATCH")
                logger.log(f"⚠️ " + "="*50)
                
                baseline_results[model_name] = {
                    'baseline_tested': False, 
                    'reason': f'Output mismatch: model={model_output_classes}, dataset={num_expected_classes}',
                    'model_classes': model_output_classes,
                    'dataset_classes': num_expected_classes,
                    'force_scratch': True,
                    'accuracy': 0.0  # Set to 0 to trigger scratch training
                }
                del model
                keras.backend.clear_session()
                gc.collect()
                continue
            
            # Predictions
            y_pred_probs = model.predict(X_test, verbose=0)
            y_pred = np.argmax(y_pred_probs, axis=1)
            
            # ==================== COMPREHENSIVE METRICS ====================
            # Basic metrics
            accuracy = accuracy_score(y_test, y_pred)
            precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
            recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
            f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
            
            # Advanced metrics
            balanced_acc = balanced_accuracy_score(y_test, y_pred)
            cohen_kappa = cohen_kappa_score(y_test, y_pred)
            try:
                mcc = matthews_corrcoef(y_test, y_pred)
            except:
                mcc = 0.0
            
            # Multi-class ROC-AUC (one-vs-rest)
            try:
                roc_auc = roc_auc_score(y_test, y_pred_probs, multi_class='ovr', average='weighted')
            except:
                roc_auc = None
            
            # Log loss
            try:
                logloss = log_loss(y_test, y_pred_probs)
            except:
                logloss = None
            
            # Per-class metrics (ALL classes)
            precision_per_class = precision_score(y_test, y_pred, average=None, zero_division=0)
            recall_per_class = recall_score(y_test, y_pred, average=None, zero_division=0)
            f1_per_class = f1_score(y_test, y_pred, average=None, zero_division=0)
            
            # Macro and Micro averages
            precision_macro = precision_score(y_test, y_pred, average='macro', zero_division=0)
            recall_macro = recall_score(y_test, y_pred, average='macro', zero_division=0)
            f1_macro = f1_score(y_test, y_pred, average='macro', zero_division=0)
            
            # Top-K accuracy
            try:
                top3_acc = top_k_accuracy_score(y_test, y_pred_probs, k=3)
                top5_acc = top_k_accuracy_score(y_test, y_pred_probs, k=5)
            except:
                top3_acc = None
                top5_acc = None
            
            # Confidence analysis
            pred_confidences = np.max(y_pred_probs, axis=1)
            avg_confidence = np.mean(pred_confidences)
            correct_mask = y_pred == y_test
            avg_conf_correct = np.mean(pred_confidences[correct_mask]) if np.any(correct_mask) else 0
            avg_conf_wrong = np.mean(pred_confidences[~correct_mask]) if np.any(~correct_mask) else 0
            
            # Confusion matrix for detailed analysis
            cm = confusion_matrix(y_test, y_pred)
            
            # Per-class accuracy
            per_class_acc = cm.diagonal() / cm.sum(axis=1) if cm.sum() > 0 else np.zeros(len(class_names))
            
            # Find most confused pairs
            cm_no_diag = cm.copy()
            np.fill_diagonal(cm_no_diag, 0)
            confused_pairs = []
            for i in range(len(class_names)):
                for j in range(len(class_names)):
                    if cm_no_diag[i, j] > 0:
                        confused_pairs.append({
                            'true': class_names[i], 
                            'pred': class_names[j], 
                            'count': int(cm_no_diag[i, j])
                        })
            confused_pairs.sort(key=lambda x: x['count'], reverse=True)
            
            # Determine if model needs scratch training
            # - Low accuracy (<30%) indicates preprocessing/data mismatch
            # - Inference-only models (SavedModel format) cannot be trained further
            force_scratch = accuracy < 0.30  # Less than 30% suggests major issue
            
            # If it's an inference-only model, note that we need scratch training
            if is_inference_only:
                logger.log(f"\n⚠️ Note: {model_name} is loaded as inference-only (SavedModel format)")
                logger.log(f"   This model CANNOT be fine-tuned - will train new model from scratch")
                force_scratch = True
            
            baseline_results[model_name] = {
                'baseline_tested': True,
                'samples_tested': len(X_test),
                'accuracy': float(accuracy),
                'balanced_accuracy': float(balanced_acc),
                'precision': float(precision),
                'recall': float(recall),
                'f1_score': float(f1),
                'precision_macro': float(precision_macro),
                'recall_macro': float(recall_macro),
                'f1_macro': float(f1_macro),
                'cohen_kappa': float(cohen_kappa),
                'matthews_corrcoef': float(mcc),
                'roc_auc': float(roc_auc) if roc_auc else None,
                'log_loss': float(logloss) if logloss else None,
                'top3_accuracy': float(top3_acc) if top3_acc else None,
                'top5_accuracy': float(top5_acc) if top5_acc else None,
                'avg_confidence': float(avg_confidence),
                'avg_conf_correct': float(avg_conf_correct),
                'avg_conf_wrong': float(avg_conf_wrong),
                'per_class_f1': {class_names[i]: float(f1_per_class[i]) for i in range(len(class_names)) if i < len(f1_per_class)},
                'per_class_accuracy': {class_names[i]: float(per_class_acc[i]) for i in range(len(class_names)) if i < len(per_class_acc)},
                'per_class_precision': {class_names[i]: float(precision_per_class[i]) for i in range(len(class_names)) if i < len(precision_per_class)},
                'per_class_recall': {class_names[i]: float(recall_per_class[i]) for i in range(len(class_names)) if i < len(recall_per_class)},
                'most_confused_pairs': confused_pairs[:20],  # Top 20 confused pairs
                'confusion_matrix': cm.tolist(),
                'y_test': y_test.tolist(),
                'y_pred': y_pred.tolist(),
                'y_pred_probs': y_pred_probs.tolist(),  # Keep for ROC curves
                'force_scratch': force_scratch
            }
            
            # ==================== DETAILED BASELINE METRICS DISPLAY ====================
            logger.log(f"\n{'='*70}")
            logger.log(f"📊 COMPREHENSIVE BASELINE RESULTS for {model_name}")
            logger.log(f"{'='*70}")
            
            logger.log(f"\n📈 OVERALL METRICS:")
            logger.log(f"   ┌───────────────────────────────────────────────────────┐")
            logger.log(f"   │ Accuracy:           {accuracy:.4f} ({accuracy*100:.2f}%)              │")
            logger.log(f"   │ Balanced Accuracy:  {balanced_acc:.4f} ({balanced_acc*100:.2f}%)              │")
            logger.log(f"   │ Precision (weighted): {precision:.4f}                         │")
            logger.log(f"   │ Recall (weighted):    {recall:.4f}                         │")
            logger.log(f"   │ F1-Score (weighted):  {f1:.4f}                         │")
            logger.log(f"   │ F1-Score (macro):     {f1_macro:.4f}                         │")
            logger.log(f"   │ Cohen's Kappa:        {cohen_kappa:.4f}                         │")
            logger.log(f"   │ Matthews CC:          {mcc:.4f}                         │")
            if roc_auc:
                logger.log(f"   │ ROC-AUC (weighted):   {roc_auc:.4f}                         │")
            if logloss:
                logger.log(f"   │ Log Loss:             {logloss:.4f}                         │")
            if top3_acc:
                logger.log(f"   │ Top-3 Accuracy:       {top3_acc:.4f} ({top3_acc*100:.2f}%)              │")
            if top5_acc:
                logger.log(f"   │ Top-5 Accuracy:       {top5_acc:.4f} ({top5_acc*100:.2f}%)              │")
            logger.log(f"   └───────────────────────────────────────────────────────┘")
            
            logger.log(f"\n🎯 CONFIDENCE ANALYSIS:")
            logger.log(f"   • Average Confidence:     {avg_confidence:.4f}")
            logger.log(f"   • Conf (Correct Preds):   {avg_conf_correct:.4f}")
            logger.log(f"   • Conf (Wrong Preds):     {avg_conf_wrong:.4f}")
            confidence_gap = avg_conf_correct - avg_conf_wrong
            logger.log(f"   • Confidence Gap:         {confidence_gap:.4f} {'✅ (good calibration)' if confidence_gap > 0.1 else '⚠️ (poor calibration)'}")
            
            # Display per-class metrics for ALL classes
            logger.log(f"\n📋 PER-CLASS METRICS (ALL {len(class_names)} CLASSES):")
            logger.log(f"   {'Class':<8} {'Accuracy':>10} {'Precision':>10} {'Recall':>10} {'F1-Score':>10} {'Support':>8}")
            logger.log(f"   {'-'*56}")
            for i, cls_name in enumerate(class_names):
                if i < len(per_class_acc):
                    support = int(cm.sum(axis=1)[i]) if i < len(cm) else 0
                    logger.log(f"   {cls_name:<8} {per_class_acc[i]:>10.4f} {precision_per_class[i]:>10.4f} {recall_per_class[i]:>10.4f} {f1_per_class[i]:>10.4f} {support:>8}")
            
            # Most confused pairs
            logger.log(f"\n🔍 MOST CONFUSED PAIRS (Top 10):")
            for pair in confused_pairs[:10]:
                logger.log(f"   {pair['true']} → {pair['pred']}: {pair['count']} mistakes")
            
            # Create and save baseline confusion matrix
            visualize_baseline_confusion_matrix(cm, class_names, model_name)
            
            # Warn if accuracy is suspiciously low (below 30% indicates class mismatch)
            if accuracy < 0.30:
                logger.log(f"\n" + "="*60)
                logger.log(f"⚠️ CRITICAL: {model_name} has very low accuracy ({accuracy*100:.1f}%)!")
                logger.log(f"="*60)
                logger.log(f"   This indicates the model was likely trained on")
                logger.log(f"   different data (e.g., 26 classes vs current {num_expected_classes} classes).")
                logger.log(f"   High confidence ({avg_confidence*100:.1f}%) with low accuracy")
                logger.log(f"   suggests PREPROCESSING MISMATCH!")
                logger.log(f"   ")
                logger.log(f"   DIAGNOSTIC: Check if model was trained with:")
                logger.log(f"   - Different preprocessing (e.g., /255 vs preprocess_input)")
                logger.log(f"   - Different image size or color space")
                logger.log(f"   - Different augmentations")
                logger.log(f"   ")
                logger.log(f"   → WILL FORCE training from SCRATCH")
                logger.log(f"="*60)
            elif accuracy < 0.7:
                logger.log(f"\n⚠️ {model_name} has below-expected accuracy ({accuracy*100:.1f}%)")
                logger.log(f"   Consider training from scratch for better results")
            
            # Cleanup
            del model, X_test, y_test
            keras.backend.clear_session()
            gc.collect()
            
        except Exception as e:
            logger.log(f"❌ Error testing {model_name}: {e}")
            baseline_results[model_name] = {'baseline_tested': False, 'reason': str(e), 'force_scratch': True}
    
    # Save baseline results
    baseline_path = os.path.join(CONFIG['metrics_path'], 'baseline_results.json')
    with open(baseline_path, 'w') as f:
        # Remove large arrays before saving and convert numpy types to native Python
        save_results = {}
        skip_keys = ['y_test', 'y_pred', 'y_pred_probs', 'confusion_matrix']  # Skip large arrays
        for name, results in baseline_results.items():
            save_results[name] = {}
            for k, v in results.items():
                if k in skip_keys:
                    continue  # Skip large arrays
                # Convert numpy types to native Python types
                if hasattr(v, 'item'):  # numpy scalar
                    save_results[name][k] = v.item()
                elif isinstance(v, (np.bool_, bool)):
                    save_results[name][k] = bool(v)
                elif isinstance(v, (np.integer, np.floating)):
                    save_results[name][k] = float(v)
                elif isinstance(v, dict):
                    # Convert dict values recursively
                    converted_dict = {}
                    for dk, dv in v.items():
                        if hasattr(dv, 'item'):
                            converted_dict[dk] = dv.item()
                        elif isinstance(dv, (np.bool_, bool)):
                            converted_dict[dk] = bool(dv)
                        elif isinstance(dv, (np.integer, np.floating)):
                            converted_dict[dk] = float(dv)
                        else:
                            converted_dict[dk] = dv
                    save_results[name][k] = converted_dict
                elif isinstance(v, list):
                    # Handle lists (e.g., most_confused_pairs)
                    save_results[name][k] = v
                else:
                    save_results[name][k] = v
        json.dump(save_results, f, indent=2)
    logger.log(f"\n💾 Baseline results saved to: {baseline_path}")
    
    # Create interactive baseline comparison visualization
    visualize_baseline_results(baseline_results, class_names)
    
    # Print summary
    logger.log(f"\n{'='*90}")
    logger.log(f"📊 BASELINE SUMMARY")
    logger.log(f"{'='*90}")
    logger.log(f"{'Model':<20} {'Status':<12} {'Accuracy':>10} {'Precision':>10} {'Recall':>10} {'F1-Score':>10}")
    logger.log(f"{'-'*90}")
    for name, results in baseline_results.items():
        if results.get('baseline_tested'):
            logger.log(f"{name:<20} {'✅ Tested':<12} {results['accuracy']*100:>9.2f}% {results['precision']:>10.4f} {results['recall']:>10.4f} {results['f1_score']:>10.4f}")
        else:
            reason = results.get('reason', 'Unknown')[:20]
            logger.log(f"{name:<20} {'❌ Failed':<12} {'N/A':>10} {'N/A':>10} {'N/A':>10} {'N/A':>10}")
            logger.log(f"   └─ Reason: {reason}")
    logger.log(f"{'='*90}")
    
    # Identify models that need retraining from scratch
    models_to_retrain = []
    for name, results in baseline_results.items():
        if results.get('baseline_tested') and results.get('accuracy', 0) < 0.5:
            models_to_retrain.append(name)
        elif not results.get('baseline_tested'):
            models_to_retrain.append(name)
    
    if models_to_retrain:
        logger.log(f"\n⚠️ MODELS RECOMMENDED FOR SCRATCH TRAINING:")
        for name in models_to_retrain:
            logger.log(f"   • {name}")
    
    return baseline_results

def visualize_baseline_results(baseline_results, class_names):
    """Create interactive visualization of baseline test results"""
    
    logger.log("\n📊 Creating interactive baseline results visualization...")
    
    # Filter tested models
    tested_models = {k: v for k, v in baseline_results.items() if v.get('baseline_tested')}
    
    if not tested_models:
        logger.log("⚠️ No models tested - skipping visualization")
        return
    
    # Create comprehensive baseline dashboard
    fig = make_subplots(
        rows=2, cols=2,
        specs=[
            [{"type": "bar"}, {"type": "scatter3d"}],
            [{"type": "bar"}, {"type": "heatmap"}]
        ],
        subplot_titles=[
            '📊 Model Metrics Comparison',
            '🎯 3D Metrics Space',
            '📈 Confidence Analysis',
            '🔥 Per-Class F1 Scores'
        ],
        vertical_spacing=0.12,
        horizontal_spacing=0.1
    )
    
    models = list(tested_models.keys())
    
    # Plot 1: All metrics comparison
    metrics = ['accuracy', 'precision', 'recall', 'f1_score']
    colors = ['#2ecc71', '#3498db', '#e74c3c', '#9b59b6']
    
    for i, metric in enumerate(metrics):
        values = [tested_models[m].get(metric, 0) for m in models]
        fig.add_trace(
            go.Bar(
                name=metric.replace('_', ' ').title(),
                x=models,
                y=values,
                marker_color=colors[i],
                text=[f'{v:.3f}' for v in values],
                textposition='outside',
                hovertemplate=f'<b>%{{x}}</b><br>{metric.title()}: %{{y:.4f}}<extra></extra>'
            ),
            row=1, col=1
        )
    
    # Plot 2: 3D scatter - accuracy vs f1 vs confidence
    accuracies = [tested_models[m].get('accuracy', 0) for m in models]
    f1_scores = [tested_models[m].get('f1_score', 0) for m in models]
    confidences = [tested_models[m].get('avg_confidence', 0) for m in models]
    
    fig.add_trace(
        go.Scatter3d(
            x=accuracies,
            y=f1_scores,
            z=confidences,
            mode='markers+text',
            marker=dict(
                size=15,
                color=accuracies,
                colorscale='RdYlGn',
                showscale=True,
                colorbar=dict(title='Accuracy', x=0.45, len=0.4, y=0.8)
            ),
            text=models,
            textposition='top center',
            hovertemplate='<b>%{text}</b><br>Accuracy: %{x:.4f}<br>F1: %{y:.4f}<br>Confidence: %{z:.4f}<extra></extra>'
        ),
        row=1, col=2
    )
    
    # Plot 3: Confidence analysis
    conf_correct = [tested_models[m].get('avg_conf_correct', 0) for m in models]
    conf_wrong = [tested_models[m].get('avg_conf_wrong', 0) for m in models]
    
    fig.add_trace(
        go.Bar(
            name='Conf (Correct)',
            x=models,
            y=conf_correct,
            marker_color='#27ae60',
            hovertemplate='<b>%{x}</b><br>Confidence on Correct: %{y:.4f}<extra></extra>'
        ),
        row=2, col=1
    )
    fig.add_trace(
        go.Bar(
            name='Conf (Wrong)',
            x=models,
            y=conf_wrong,
            marker_color='#e74c3c',
            hovertemplate='<b>%{x}</b><br>Confidence on Wrong: %{y:.4f}<extra></extra>'
        ),
        row=2, col=1
    )
    
    # Plot 4: Per-class F1 heatmap (for first model with per-class data)
    for model_name, results in tested_models.items():
        if 'per_class_f1' in results:
            per_class = results['per_class_f1']
            classes = list(per_class.keys())
            f1_values = list(per_class.values())
            
            # Create matrix for all models
            all_f1 = []
            for m in models:
                if 'per_class_f1' in tested_models[m]:
                    m_f1 = [tested_models[m]['per_class_f1'].get(c, 0) for c in classes]
                else:
                    m_f1 = [0] * len(classes)
                all_f1.append(m_f1)
            
            fig.add_trace(
                go.Heatmap(
                    z=all_f1,
                    x=classes,
                    y=models,
                    colorscale='RdYlGn',
                    showscale=True,
                    colorbar=dict(title='F1', x=1.02, len=0.4, y=0.2),
                    hovertemplate='<b>%{y}</b><br>Class %{x}: F1=%{z:.4f}<extra></extra>'
                ),
                row=2, col=2
            )
            break
    
    # Update layout
    fig.update_layout(
        title=dict(
            text='📊 Baseline Test Results Dashboard (Interactive)',
            font=dict(size=22, color='#333'),
            x=0.5
        ),
        height=900,
        template='plotly_white',
        barmode='group',
        showlegend=True,
        legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1)
    )
    
    # Update 3D scene
    fig.update_scenes(
        xaxis_title='Accuracy',
        yaxis_title='F1 Score',
        zaxis_title='Confidence',
        camera=dict(eye=dict(x=1.5, y=1.5, z=1.2))
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], 'baseline_results_dashboard.png')
    save_and_show_plotly(fig, save_path, 'Baseline Results Dashboard')

# ==================== VISUALIZATION HELPERS ====================
def visualize_augmentation_preview(train_paths, train_labels, aug_pipeline, model_name):
    """Preview augmentation effects on sample images"""
    
    logger.log(f"\n📊 Creating augmentation preview for {model_name}...")
    
    # Select 4 random samples
    np.random.seed(42)
    sample_indices = np.random.choice(len(train_paths), min(4, len(train_paths)), replace=False)
    
    fig, axes = plt.subplots(4, 5, figsize=(15, 12))
    
    for row, idx in enumerate(sample_indices):
        img_path = train_paths[idx]
        img = cv2.imread(img_path)
        if img is None:
            continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, CONFIG['img_size'])
        
        # Original
        axes[row, 0].imshow(img)
        axes[row, 0].set_title('Original' if row == 0 else '', fontsize=10)
        axes[row, 0].axis('off')
        
        # Augmented versions
        if aug_pipeline:
            for col in range(1, 5):
                try:
                    aug_img = aug_pipeline(image=img)['image']
                    axes[row, col].imshow(aug_img)
                except:
                    axes[row, col].imshow(img)
                axes[row, col].set_title(f'Aug {col}' if row == 0 else '', fontsize=10)
                axes[row, col].axis('off')
        else:
            for col in range(1, 5):
                axes[row, col].imshow(img)
                axes[row, col].axis('off')
    
    plt.suptitle(f'{model_name} - Augmentation Preview', fontsize=14, fontweight='bold')
    plt.tight_layout()
    save_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_augmentation_preview.png')
    save_and_show_plot(save_path, f'{model_name} Augmentation Preview')

def plot_training_history_3d(history, model_name):
    """Create INTERACTIVE 3D visualization of training history with Plotly"""
    
    logger.log(f"\n📊 Creating interactive 3D training surface for {model_name}...")
    
    epochs = list(range(1, len(history.history['accuracy']) + 1))
    train_acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    train_loss = history.history['loss']
    val_loss = history.history['val_loss']
    
    # Create subplot with two 3D charts
    fig = make_subplots(
        rows=1, cols=2,
        specs=[[{"type": "scene"}, {"type": "scene"}]],
        subplot_titles=['📈 Accuracy Progress', '📉 Loss Progress']
    )
    
    # Plot 1: 3D Accuracy Trajectories
    fig.add_trace(
        go.Scatter3d(
            x=epochs, y=train_acc, z=[0]*len(epochs),
            mode='lines+markers',
            marker=dict(size=5, color='blue'),
            line=dict(color='blue', width=3),
            name='Train Accuracy',
            hovertemplate='Epoch %{x}<br>Train Acc: %{y:.4f}<extra></extra>'
        ),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter3d(
            x=epochs, y=val_acc, z=[1]*len(epochs),
            mode='lines+markers',
            marker=dict(size=5, color='red'),
            line=dict(color='red', width=3),
            name='Val Accuracy',
            hovertemplate='Epoch %{x}<br>Val Acc: %{y:.4f}<extra></extra>'
        ),
        row=1, col=1
    )
    
    # Plot 2: 3D Loss Trajectories
    fig.add_trace(
        go.Scatter3d(
            x=epochs, y=train_loss, z=[0]*len(epochs),
            mode='lines+markers',
            marker=dict(size=5, color='blue'),
            line=dict(color='blue', width=3),
            name='Train Loss',
            hovertemplate='Epoch %{x}<br>Train Loss: %{y:.4f}<extra></extra>',
            showlegend=False
        ),
        row=1, col=2
    )
    fig.add_trace(
        go.Scatter3d(
            x=epochs, y=val_loss, z=[1]*len(epochs),
            mode='lines+markers',
            marker=dict(size=5, color='red'),
            line=dict(color='red', width=3),
            name='Val Loss',
            hovertemplate='Epoch %{x}<br>Val Loss: %{y:.4f}<extra></extra>',
            showlegend=False
        ),
        row=1, col=2
    )
    
    # Update scenes
    fig.update_scenes(
        xaxis_title='Epoch',
        yaxis_title='Accuracy',
        zaxis_title='Type (0=Train, 1=Val)',
        camera=dict(eye=dict(x=1.5, y=1.5, z=1.0)),
        row=1, col=1
    )
    fig.update_scenes(
        xaxis_title='Epoch',
        yaxis_title='Loss',
        zaxis_title='Type (0=Train, 1=Val)',
        camera=dict(eye=dict(x=1.5, y=1.5, z=1.0)),
        row=1, col=2
    )
    
    fig.update_layout(
        title=dict(
            text=f'📊 {model_name} - 3D Training Visualization (Rotate & Zoom!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        height=600,
        template='plotly_white'
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_training_3d.png')
    save_and_show_plotly(fig, save_path, f'{model_name} 3D Training')

def visualize_feature_embeddings_3d(model, model_name, test_paths, test_labels, class_names):
    """Extract features and visualize in INTERACTIVE 3D using t-SNE with Plotly"""
    
    logger.log(f"\n📊 Creating interactive 3D feature embeddings for {model_name}...")
    
    # Create feature extractor (remove last dense layer)
    try:
        feature_extractor = keras.Model(
            inputs=model.input,
            outputs=model.layers[-3].output  # Get features before final dense
        )
    except:
        logger.log("⚠️ Could not create feature extractor, skipping embedding visualization")
        return
    
    # Set preprocessing function
    if model_name == 'EfficientNetV2B3':
        preprocess_fn = effnet_preprocess
    elif model_name == 'ResNet50':
        preprocess_fn = resnet_preprocess
    elif model_name == 'InceptionV3':
        preprocess_fn = inception_preprocess
    else:
        preprocess_fn = None
    
    # Use stratified sampling to ensure ALL classes are represented
    # Sample up to 2000 samples for good visualization while keeping all classes
    max_samples_per_class = max(20, min(100, len(test_paths) // len(class_names)))
    
    # Group samples by class
    class_samples = {i: [] for i in range(len(class_names))}
    for idx, label in enumerate(test_labels):
        class_samples[label].append(idx)
    
    # Sample from each class
    sample_indices = []
    for class_idx, indices in class_samples.items():
        n_samples = min(max_samples_per_class, len(indices))
        if n_samples > 0:
            np.random.seed(42 + class_idx)
            sampled = np.random.choice(indices, n_samples, replace=False)
            sample_indices.extend(sampled)
    
    logger.log(f"   Sampling {len(sample_indices)} samples ({max_samples_per_class} per class) for embedding visualization")
    
    X_sample = []
    y_sample = []
    
    for idx in sample_indices:
        img = cv2.imread(test_paths[idx])
        if img is None:
            continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, CONFIG['img_size'])
        
        if preprocess_fn:
            img = preprocess_fn(img.astype(np.float32))
        else:
            img = img.astype(np.float32) / 255.0
        
        X_sample.append(img)
        y_sample.append(test_labels[idx])
    
    X_sample = np.array(X_sample)
    y_sample = np.array(y_sample)
    
    if len(X_sample) < 50:
        logger.log("⚠️ Not enough samples for embedding visualization")
        return
    
    # Extract features
    features = feature_extractor.predict(X_sample, verbose=0)
    
    # Flatten if needed
    if len(features.shape) > 2:
        features = features.reshape(features.shape[0], -1)
    
    # Apply PCA first for dimensionality reduction
    pca = PCA(n_components=50)
    features_pca = pca.fit_transform(features)
    
    # Apply t-SNE for 3D
    logger.log("  Computing t-SNE (this may take a moment)...")
    tsne = TSNE(n_components=3, random_state=42, perplexity=30, n_iter=1000)
    features_3d = tsne.fit_transform(features_pca)
    
    # Create INTERACTIVE 3D scatter plot with Plotly
    unique_classes = np.unique(y_sample)
    
    # Generate color palette
    colors = px.colors.qualitative.Dark24 + px.colors.qualitative.Light24
    
    fig = go.Figure()
    
    for i, class_idx in enumerate(unique_classes):
        mask = y_sample == class_idx
        fig.add_trace(go.Scatter3d(
            x=features_3d[mask, 0],
            y=features_3d[mask, 1],
            z=features_3d[mask, 2],
            mode='markers',
            name=class_names[class_idx],
            marker=dict(
                size=5,
                color=colors[i % len(colors)],
                opacity=0.7,
                line=dict(width=1, color='darkgray')
            ),
            hovertemplate=f'<b>Class: {class_names[class_idx]}</b><br>t-SNE1: %{{x:.2f}}<br>t-SNE2: %{{y:.2f}}<br>t-SNE3: %{{z:.2f}}<extra></extra>'
        ))
    
    fig.update_layout(
        title=dict(
            text=f'🎯 {model_name} - Feature Embeddings t-SNE 3D (Rotate & Explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis_title='t-SNE 1',
            yaxis_title='t-SNE 2',
            zaxis_title='t-SNE 3',
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.2))
        ),
        legend=dict(
            itemsizing='constant',
            font=dict(size=10),
            yanchor='top', y=0.99,
            xanchor='left', x=1.02
        ),
        height=800,
        template='plotly_white',
        margin=dict(l=0, r=150, t=50, b=0)
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_embeddings_3d.png')
    save_and_show_plotly(fig, save_path, f'{model_name} 3D Embeddings')
    
    # Note: Skipping 2D t-SNE - 3D version is more informative

# ==================== TRAINING ====================
class DetailedMetricsCallback(keras.callbacks.Callback):
    """Custom callback to print detailed metrics during training"""
    
    def __init__(self, model_name, total_epochs):
        super().__init__()
        self.model_name = model_name
        self.total_epochs = total_epochs
        self.best_val_acc = 0
        self.best_epoch = 0
        self.start_time = None
        self.epoch_times = []
    
    def on_train_begin(self, logs=None):
        self.start_time = time.time()
        print(f"\n{'='*80}")
        print(f"🚀 TRAINING STARTED: {self.model_name}")
        print(f"{'='*80}")
    
    def on_epoch_begin(self, epoch, logs=None):
        self.epoch_start = time.time()
        print(f"\n📍 Epoch {epoch+1}/{self.total_epochs}")
    
    def on_epoch_end(self, epoch, logs=None):
        epoch_time = time.time() - self.epoch_start
        self.epoch_times.append(epoch_time)
        
        # Get metrics
        train_acc = logs.get('accuracy', 0)
        val_acc = logs.get('val_accuracy', 0)
        train_loss = logs.get('loss', 0)
        val_loss = logs.get('val_loss', 0)
        lr = float(keras.backend.get_value(self.model.optimizer.learning_rate))
        
        # Track best
        if val_acc > self.best_val_acc:
            self.best_val_acc = val_acc
            self.best_epoch = epoch + 1
            best_marker = " ⭐ NEW BEST!"
        else:
            best_marker = ""
        
        # Calculate ETA
        avg_epoch_time = sum(self.epoch_times) / len(self.epoch_times)
        remaining_epochs = self.total_epochs - (epoch + 1)
        eta_seconds = avg_epoch_time * remaining_epochs
        eta_str = f"{eta_seconds/60:.1f}min" if eta_seconds < 3600 else f"{eta_seconds/3600:.1f}hr"
        
        # Print detailed metrics
        print(f"   ┌{'─'*60}┐")
        print(f"   │ Train Acc: {train_acc*100:6.2f}%  │  Val Acc: {val_acc*100:6.2f}%{best_marker:>15} │")
        print(f"   │ Train Loss: {train_loss:6.4f}   │  Val Loss: {val_loss:6.4f}              │")
        print(f"   │ LR: {lr:.2e}      │  Time: {epoch_time:.1f}s  │  ETA: {eta_str:>8} │")
        print(f"   │ Best Val Acc: {self.best_val_acc*100:.2f}% @ Epoch {self.best_epoch:2d}                  │")
        print(f"   └{'─'*60}┘")
    
    def on_train_end(self, logs=None):
        total_time = time.time() - self.start_time
        print(f"\n{'='*80}")
        print(f"✅ TRAINING COMPLETE: {self.model_name}")
        print(f"{'='*80}")
        print(f"   Total time: {total_time/60:.1f} minutes")
        print(f"   Best val accuracy: {self.best_val_acc*100:.2f}% @ Epoch {self.best_epoch}")
        print(f"   Average epoch time: {sum(self.epoch_times)/len(self.epoch_times):.1f}s")
        print(f"{'='*80}\n")

def check_existing_model(model_name):
    """Check if existing model can be loaded for resume training"""
    
    model_path = CONFIG['existing_models'].get(model_name)
    
    if model_path and os.path.exists(model_path):
        try:
            model = keras.models.load_model(model_path)
            logger.log(f"✅ Found existing model: {model_path}")
            return model, model_path
        except Exception as e:
            logger.log(f"⚠️ Failed to load {model_path}: {e}")
    
    return None, None

def train_model(model_name, train_paths, train_labels, val_paths, val_labels, 
                num_classes, class_names, force_scratch=False, training_mode='AUTO'):
    """Train or resume training for a model with intelligent mode selection
    
    Args:
        force_scratch: If True, ignore existing model and train from scratch
        training_mode: One of 'AUTO', 'SCRATCH', 'RESUME', 'FINE_TUNE'
            - AUTO: Automatically decide based on existing model
            - SCRATCH: Train new model from scratch
            - RESUME: Resume training from checkpoint (moderate epochs, low LR)
            - FINE_TUNE: Light fine-tuning (few epochs, very low LR)
    """
    
    logger.section(f"TRAINING: {model_name}")
    logger.log(f"🎯 Training mode: {training_mode}")
    
    # Handle force_scratch legacy parameter
    if force_scratch:
        training_mode = 'SCRATCH'
    
    # Check for existing model (unless forced to scratch)
    if training_mode == 'SCRATCH':
        logger.log(f"🆕 FORCED training from SCRATCH (ignoring existing model)")
        existing_model, model_path = None, None
    else:
        existing_model, model_path = check_existing_model(model_name)
    
    # Determine training parameters based on mode
    if training_mode == 'FINE_TUNE' and existing_model:
        # Fine-tuning: very few epochs, minimal learning rate
        logger.log(f"🔧 FINE-TUNING from: {model_path}")
        model = existing_model
        epochs = CONFIG.get('fine_tune_epochs', 5)  # Very few epochs
        initial_lr = 5e-6  # Very low LR for fine-tuning
        is_resume = True
        logger.log(f"   • Fine-tune epochs: {epochs}")
        logger.log(f"   • Fine-tune LR: {initial_lr}")
    elif training_mode == 'RESUME' and existing_model:
        # Resume: moderate epochs, low learning rate
        logger.log(f"🔄 RESUMING training from: {model_path}")
        model = existing_model
        epochs = CONFIG['resume_epochs']
        initial_lr = 1e-5  # Low LR for continued training
        is_resume = True
    elif existing_model and training_mode == 'AUTO':
        # Auto mode with existing model - treat as resume
        logger.log(f"🔄 AUTO mode: RESUMING from: {model_path}")
        model = existing_model
        epochs = CONFIG['resume_epochs']
        initial_lr = 1e-5
        is_resume = True
    else:
        # Scratch training (new model or forced)
        logger.log(f"🆕 Training from SCRATCH")
        if training_mode not in ['SCRATCH', 'AUTO']:
            logger.log(f"⚠️ No existing model found - falling back to SCRATCH")
        if training_mode == 'AUTO':
            logger.log(f"⚠️ No existing model found at: {CONFIG['existing_models'].get(model_name, 'Not configured')}")
        model = MODEL_BUILDERS[model_name](num_classes)
        epochs = CONFIG['epochs']
        initial_lr = 1e-3
        is_resume = False
    
    # Compile model
    optimizer = keras.optimizers.Adam(learning_rate=initial_lr)
    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    
    # Print detailed model info
    total_params = model.count_params()
    trainable_params = sum([tf.size(w).numpy() for w in model.trainable_weights])
    non_trainable_params = total_params - trainable_params
    
    logger.log(f"\n" + "=" * 60)
    logger.log(f"📊 MODEL ARCHITECTURE: {model.name}")
    logger.log(f"=" * 60)
    logger.log(f"📊 Total parameters:        {total_params:,}")
    logger.log(f"📊 Trainable parameters:    {trainable_params:,}")
    logger.log(f"📊 Non-trainable params:    {non_trainable_params:,}")
    logger.log(f"📊 Trainable ratio:         {trainable_params/total_params*100:.1f}%")
    logger.log(f"=" * 60)
    logger.log(f"📊 Epochs:                  {epochs}")
    logger.log(f"📊 Initial LR:              {initial_lr}")
    logger.log(f"📊 Batch size:              {CONFIG['batch_size']}")
    logger.log(f"📊 Training samples:        {len(train_paths)}")
    logger.log(f"📊 Validation samples:      {len(val_paths)}")
    logger.log(f"📊 Steps per epoch:         {len(train_paths) // CONFIG['batch_size']}")
    logger.log(f"=" * 60)
    
    # Print layer summary to terminal
    logger.log("\n📋 Layer Summary:")
    logger.log("-" * 60)
    for i, layer in enumerate(model.layers[-10:]):  # Last 10 layers
        trainable_str = "✓" if layer.trainable else "✗"
        # Safely get output shape (TF 2.18+ compatibility)
        try:
            out_shape = layer.output.shape if hasattr(layer, 'output') else 'N/A'
        except:
            out_shape = 'N/A'
        logger.log(f"  [{trainable_str}] {layer.name}: {out_shape}")
    logger.log("-" * 60)
    
    # Create data generators
    aug_pipeline = get_augmentation_pipeline(model_name)
    
    # Preview augmentation
    visualize_augmentation_preview(train_paths, train_labels, aug_pipeline, model_name)
    
    train_gen = AugmentedDataGenerator(
        train_paths, train_labels, CONFIG['batch_size'], model_name,
        augmentation=aug_pipeline, shuffle=True
    )
    
    val_gen = AugmentedDataGenerator(
        val_paths, val_labels, CONFIG['batch_size'], model_name,
        augmentation=None, shuffle=False
    )
    
    # Callbacks
    callbacks = [
        DetailedMetricsCallback(model_name, epochs),  # Custom detailed metrics
        ModelCheckpoint(
            os.path.join(CONFIG['models_path'], f'{model_name}_best.keras'),
            save_best_only=True, monitor='val_accuracy', mode='max', verbose=0
        ),
        EarlyStopping(
            monitor='val_accuracy', patience=8, restore_best_weights=True, verbose=1
        ),
        ReduceLROnPlateau(
            monitor='val_loss', factor=0.3, patience=4, min_lr=1e-7, verbose=1
        ),
        keras.callbacks.TerminateOnNaN()
    ]
    
    # Train
    logger.log(f"\n🚀 Starting training...")
    logger.log(f"   Using GPU: {HAS_GPU}")
    start_time = time.time()
    
    try:
        history = model.fit(
            train_gen,
            validation_data=val_gen,
            epochs=epochs,
            callbacks=callbacks,
            verbose=1
        )
    except Exception as e:
        logger.log(f"❌ Training failed: {e}")
        # Save crash model
        crash_path = os.path.join(CONFIG['models_path'], f'{model_name}_crash.keras')
        try:
            model.save(crash_path)
            logger.log(f"💾 Crash model saved to: {crash_path}")
        except:
            pass
        raise
    
    training_time = time.time() - start_time
    
    # Save final model
    final_path = os.path.join(CONFIG['models_path'], f'{model_name}_final.keras')
    model.save(final_path)
    logger.log(f"\n💾 Final model saved to: {final_path}")
    logger.log(f"⏱️ Training time: {training_time/60:.1f} minutes")
    
    # Plot training history
    plot_training_history(history, model_name)
    
    return model, history

def plot_training_history(history, model_name):
    """Plot and save INTERACTIVE 3D training history with Plotly"""
    
    # Print final metrics to terminal
    final_train_acc = history.history['accuracy'][-1]
    final_val_acc = history.history['val_accuracy'][-1]
    best_val_acc = max(history.history['val_accuracy'])
    best_epoch = history.history['val_accuracy'].index(best_val_acc) + 1
    
    logger.log(f"\n📈 Training Summary for {model_name}:")
    logger.log(f"   Final Train Accuracy:    {final_train_acc:.4f} ({final_train_acc*100:.2f}%)")
    logger.log(f"   Final Val Accuracy:      {final_val_acc:.4f} ({final_val_acc*100:.2f}%)")
    logger.log(f"   Best Val Accuracy:       {best_val_acc:.4f} ({best_val_acc*100:.2f}%) @ Epoch {best_epoch}")
    logger.log(f"   Final Train Loss:        {history.history['loss'][-1]:.4f}")
    logger.log(f"   Final Val Loss:          {history.history['val_loss'][-1]:.4f}")
    
    epochs = list(range(1, len(history.history['accuracy']) + 1))
    train_acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    train_loss = history.history['loss']
    val_loss = history.history['val_loss']
    
    # Create 3D training history visualization
    fig = go.Figure()
    
    # 3D trajectory: X = Epoch, Y = Loss, Z = Accuracy
    # Training trajectory (blue)
    fig.add_trace(go.Scatter3d(
        x=epochs,
        y=train_loss,
        z=train_acc,
        mode='lines+markers',
        name='Training',
        line=dict(color='#3498db', width=4),
        marker=dict(size=5, color='#3498db'),
        hovertemplate='Epoch: %{x}<br>Loss: %{y:.4f}<br>Accuracy: %{z:.4f}<extra></extra>'
    ))
    
    # Validation trajectory (red)
    fig.add_trace(go.Scatter3d(
        x=epochs,
        y=val_loss,
        z=val_acc,
        mode='lines+markers',
        name='Validation',
        line=dict(color='#e74c3c', width=4),
        marker=dict(size=5, color='#e74c3c'),
        hovertemplate='Epoch: %{x}<br>Loss: %{y:.4f}<br>Accuracy: %{z:.4f}<extra></extra>'
    ))
    
    # Mark best epoch with a star
    best_idx = best_epoch - 1
    fig.add_trace(go.Scatter3d(
        x=[best_epoch],
        y=[val_loss[best_idx]],
        z=[best_val_acc],
        mode='markers',
        name=f'Best: {best_val_acc:.4f}',
        marker=dict(size=15, color='gold', symbol='diamond', 
                   line=dict(width=2, color='black')),
        hovertemplate=f'🏆 Best Epoch {best_epoch}<br>Val Loss: {val_loss[best_idx]:.4f}<br>Val Acc: {best_val_acc:.4f}<extra></extra>'
    ))
    
    # Add projection lines to show progress
    # Epoch floor (where accuracy and loss values project)
    for i in range(0, len(epochs), max(1, len(epochs)//10)):
        fig.add_trace(go.Scatter3d(
            x=[epochs[i], epochs[i]],
            y=[train_loss[i], val_loss[i]],
            z=[0, 0],
            mode='lines',
            line=dict(color='gray', width=1, dash='dot'),
            showlegend=False,
            hoverinfo='skip'
        ))
    
    fig.update_layout(
        title=dict(
            text=f'📊 {model_name} - 3D Training Journey (Rotate to explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(title='Epoch'),
            yaxis=dict(title='Loss'),
            zaxis=dict(title='Accuracy', range=[0, 1]),
            camera=dict(eye=dict(x=1.5, y=-1.5, z=1.0))
        ),
        height=700,
        template='plotly_white'
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_training_history.png')
    save_and_show_plotly(fig, save_path, f'{model_name} 3D Training History')
    
    # Save history data
    history_data = {
        'accuracy': [float(x) for x in history.history['accuracy']],
        'val_accuracy': [float(x) for x in history.history['val_accuracy']],
        'loss': [float(x) for x in history.history['loss']],
        'val_loss': [float(x) for x in history.history['val_loss']],
        'best_val_accuracy': float(best_val_acc),
        'best_epoch': best_epoch
    }
    
    history_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_history.json')
    with open(history_path, 'w') as f:
        json.dump(history_data, f, indent=2)

# ==================== EVALUATION ====================
def evaluate_model(model, model_name, test_paths, test_labels, class_names):
    """Comprehensive model evaluation with ALL metrics"""
    
    logger.section(f"EVALUATING: {model_name}")
    
    # Set preprocessing function
    if model_name == 'EfficientNetV2B3':
        preprocess_fn = effnet_preprocess
    elif model_name == 'ResNet50':
        preprocess_fn = resnet_preprocess
    elif model_name == 'InceptionV3':
        preprocess_fn = inception_preprocess
    else:
        preprocess_fn = None
    
    logger.log(f"   Using preprocessing: {preprocess_fn.__name__ if preprocess_fn else 'None (using /255.0)'}")
    
    # Load ALL test data
    X_test, y_test = [], []
    logger.log(f"   Loading ALL {len(test_paths)} test samples...")
    
    for path, label in zip(test_paths, test_labels):
        img = cv2.imread(path)
        if img is None:
            continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, CONFIG['img_size'])
        
        if preprocess_fn:
            img = preprocess_fn(img.astype(np.float32))
        else:
            img = img.astype(np.float32) / 255.0
        
        X_test.append(img)
        y_test.append(label)
    
    X_test = np.array(X_test)
    y_test = np.array(y_test)
    
    logger.log(f"📊 Test samples: {len(X_test)}")
    logger.log(f"   Data shape: {X_test.shape}")
    logger.log(f"   Data range: [{X_test.min():.2f}, {X_test.max():.2f}]")
    
    if len(X_test) == 0:
        logger.log("⚠️ No test samples - skipping evaluation")
        return None
    
    # Predictions
    y_pred_probs = model.predict(X_test, verbose=0)
    y_pred = np.argmax(y_pred_probs, axis=1)
    
    # ==================== COMPREHENSIVE METRICS ====================
    # Basic metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
    
    # Advanced metrics
    balanced_acc = balanced_accuracy_score(y_test, y_pred)
    cohen_kappa = cohen_kappa_score(y_test, y_pred)
    try:
        mcc = matthews_corrcoef(y_test, y_pred)
    except:
        mcc = 0.0
    
    # Multi-class ROC-AUC
    try:
        roc_auc = roc_auc_score(y_test, y_pred_probs, multi_class='ovr', average='weighted')
    except:
        roc_auc = None
    
    # Log loss
    try:
        logloss = log_loss(y_test, y_pred_probs)
    except:
        logloss = None
    
    # Per-class metrics (ALL classes)
    precision_per_class = precision_score(y_test, y_pred, average=None, zero_division=0)
    recall_per_class = recall_score(y_test, y_pred, average=None, zero_division=0)
    f1_per_class = f1_score(y_test, y_pred, average=None, zero_division=0)
    
    # Macro averages
    precision_macro = precision_score(y_test, y_pred, average='macro', zero_division=0)
    recall_macro = recall_score(y_test, y_pred, average='macro', zero_division=0)
    f1_macro = f1_score(y_test, y_pred, average='macro', zero_division=0)
    
    # Top-K accuracy
    try:
        top3_acc = top_k_accuracy_score(y_test, y_pred_probs, k=3)
        top5_acc = top_k_accuracy_score(y_test, y_pred_probs, k=5)
    except:
        top3_acc = None
        top5_acc = None
    
    # Confidence analysis
    pred_confidences = np.max(y_pred_probs, axis=1)
    avg_confidence = np.mean(pred_confidences)
    correct_mask = y_pred == y_test
    avg_conf_correct = np.mean(pred_confidences[correct_mask]) if np.any(correct_mask) else 0
    avg_conf_wrong = np.mean(pred_confidences[~correct_mask]) if np.any(~correct_mask) else 0
    
    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred)
    per_class_acc = cm.diagonal() / cm.sum(axis=1) if cm.sum() > 0 else np.zeros(len(class_names))
    
    # Most confused pairs
    cm_no_diag = cm.copy()
    np.fill_diagonal(cm_no_diag, 0)
    confused_pairs = []
    for i in range(len(class_names)):
        for j in range(len(class_names)):
            if cm_no_diag[i, j] > 0:
                confused_pairs.append({
                    'true': class_names[i], 
                    'pred': class_names[j], 
                    'count': int(cm_no_diag[i, j])
                })
    confused_pairs.sort(key=lambda x: x['count'], reverse=True)
    
    # ==================== DISPLAY COMPREHENSIVE RESULTS ====================
    logger.log(f"\n{'='*70}")
    logger.log(f"📊 COMPREHENSIVE EVALUATION RESULTS for {model_name}")
    logger.log(f"{'='*70}")
    
    logger.log(f"\n📈 OVERALL METRICS:")
    logger.log(f"   ┌───────────────────────────────────────────────────────┐")
    logger.log(f"   │ Accuracy:             {accuracy:.4f} ({accuracy*100:.2f}%)            │")
    logger.log(f"   │ Balanced Accuracy:    {balanced_acc:.4f} ({balanced_acc*100:.2f}%)            │")
    logger.log(f"   │ Precision (weighted): {precision:.4f}                       │")
    logger.log(f"   │ Recall (weighted):    {recall:.4f}                       │")
    logger.log(f"   │ F1-Score (weighted):  {f1:.4f}                       │")
    logger.log(f"   │ F1-Score (macro):     {f1_macro:.4f}                       │")
    logger.log(f"   │ Cohen's Kappa:        {cohen_kappa:.4f}                       │")
    logger.log(f"   │ Matthews CC:          {mcc:.4f}                       │")
    if roc_auc:
        logger.log(f"   │ ROC-AUC (weighted):   {roc_auc:.4f}                       │")
    if logloss:
        logger.log(f"   │ Log Loss:             {logloss:.4f}                       │")
    if top3_acc:
        logger.log(f"   │ Top-3 Accuracy:       {top3_acc:.4f} ({top3_acc*100:.2f}%)            │")
    if top5_acc:
        logger.log(f"   │ Top-5 Accuracy:       {top5_acc:.4f} ({top5_acc*100:.2f}%)            │")
    logger.log(f"   └───────────────────────────────────────────────────────┘")
    
    logger.log(f"\n🎯 CONFIDENCE ANALYSIS:")
    logger.log(f"   • Average Confidence:     {avg_confidence:.4f}")
    logger.log(f"   • Conf (Correct Preds):   {avg_conf_correct:.4f}")
    logger.log(f"   • Conf (Wrong Preds):     {avg_conf_wrong:.4f}")
    confidence_gap = avg_conf_correct - avg_conf_wrong
    logger.log(f"   • Confidence Gap:         {confidence_gap:.4f} {'✅ (well calibrated)' if confidence_gap > 0.1 else '⚠️ (needs calibration)'}")
    
    # Classification report (shows ALL classes)
    logger.log(f"\n📋 CLASSIFICATION REPORT (ALL {len(class_names)} CLASSES):")
    report = classification_report(y_test, y_pred, target_names=class_names, digits=4)
    logger.log(report)
    
    # Per-class metrics table
    logger.log(f"\n📊 PER-CLASS METRICS:")
    logger.log(f"   {'Class':<8} {'Accuracy':>10} {'Precision':>10} {'Recall':>10} {'F1-Score':>10} {'Support':>8}")
    logger.log(f"   {'-'*56}")
    for i, cls_name in enumerate(class_names):
        if i < len(per_class_acc):
            support = int(cm.sum(axis=1)[i]) if i < len(cm) else 0
            logger.log(f"   {cls_name:<8} {per_class_acc[i]:>10.4f} {precision_per_class[i]:>10.4f} {recall_per_class[i]:>10.4f} {f1_per_class[i]:>10.4f} {support:>8}")
    
    # Most confused pairs
    logger.log(f"\n🔍 MOST CONFUSED PAIRS (Top 10):")
    for pair in confused_pairs[:10]:
        logger.log(f"   {pair['true']} → {pair['pred']}: {pair['count']} mistakes")
    
    # Save detailed report
    report_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_classification_report.txt')
    with open(report_path, 'w') as f:
        f.write(f"{'='*70}\n")
        f.write(f"Model: {model_name}\n")
        f.write(f"{'='*70}\n\n")
        f.write(f"OVERALL METRICS:\n")
        f.write(f"  Accuracy: {accuracy:.4f}\n")
        f.write(f"  Balanced Accuracy: {balanced_acc:.4f}\n")
        f.write(f"  Precision (weighted): {precision:.4f}\n")
        f.write(f"  Recall (weighted): {recall:.4f}\n")
        f.write(f"  F1-Score (weighted): {f1:.4f}\n")
        f.write(f"  F1-Score (macro): {f1_macro:.4f}\n")
        f.write(f"  Cohen's Kappa: {cohen_kappa:.4f}\n")
        f.write(f"  Matthews CC: {mcc:.4f}\n")
        if roc_auc:
            f.write(f"  ROC-AUC: {roc_auc:.4f}\n")
        if logloss:
            f.write(f"  Log Loss: {logloss:.4f}\n")
        if top3_acc:
            f.write(f"  Top-3 Accuracy: {top3_acc:.4f}\n")
        if top5_acc:
            f.write(f"  Top-5 Accuracy: {top5_acc:.4f}\n")
        f.write(f"\nCLASSIFICATION REPORT:\n")
        f.write(f"{report}\n")
        f.write(f"\nMOST CONFUSED PAIRS:\n")
        for pair in confused_pairs[:20]:
            f.write(f"  {pair['true']} → {pair['pred']}: {pair['count']}\n")
    
    # ==================== 3D CONFUSION MATRIX VISUALIZATIONS ====================
    # Restore original cm (diagonal was modified for confusion analysis)
    cm = confusion_matrix(y_test, y_pred)
    n_classes = len(class_names)
    
    # Create 3D Surface confusion matrix
    logger.log(f"\n📊 Creating 3D confusion matrix for {model_name}...")
    
    fig_cm = make_subplots(
        rows=1, cols=2,
        specs=[[{"type": "surface"}, {"type": "surface"}]],
        subplot_titles=['Raw Confusion Matrix', 'Normalized Confusion Matrix'],
        horizontal_spacing=0.05
    )
    
    # 3D Surface for raw confusion matrix
    fig_cm.add_trace(
        go.Surface(
            z=cm,
            x=list(range(n_classes)),
            y=list(range(n_classes)),
            colorscale='Blues',
            showscale=True,
            colorbar=dict(title='Count', x=0.45, len=0.8),
            hovertemplate='True: %{y}<br>Pred: %{x}<br>Count: %{z}<extra></extra>',
            contours=dict(
                z=dict(show=True, usecolormap=True, highlightcolor="white", project_z=True)
            )
        ),
        row=1, col=1
    )
    
    # Normalized confusion matrix
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    cm_normalized = np.nan_to_num(cm_normalized)
    
    fig_cm.add_trace(
        go.Surface(
            z=cm_normalized * 100,
            x=list(range(n_classes)),
            y=list(range(n_classes)),
            colorscale='RdYlGn',
            showscale=True,
            colorbar=dict(title='%', x=1.0, len=0.8),
            hovertemplate='True: %{y}<br>Pred: %{x}<br>Rate: %{z:.1f}%<extra></extra>',
            contours=dict(
                z=dict(show=True, usecolormap=True, highlightcolor="white", project_z=True)
            )
        ),
        row=1, col=2
    )
    
    # Update both scenes
    for col in [1, 2]:
        fig_cm.update_scenes(
            dict(
                xaxis=dict(title='Predicted', tickvals=list(range(n_classes)), ticktext=class_names, tickangle=45),
                yaxis=dict(title='True', tickvals=list(range(n_classes)), ticktext=class_names, tickangle=-45),
                zaxis=dict(title='Count' if col == 1 else 'Rate %'),
                camera=dict(eye=dict(x=1.5, y=-1.5, z=1.0))
            ),
            row=1, col=col
        )
    
    fig_cm.update_layout(
        title=dict(
            text=f'🎯 {model_name} - 3D Confusion Matrix (Rotate to explore!)',
            font=dict(size=18, color='#333'),
            x=0.5
        ),
        height=700,
        template='plotly_white'
    )
    
    cm_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_confusion_matrix.png')
    save_and_show_plotly(fig_cm, cm_path, f'{model_name} 3D Confusion Matrix')

    # ==================== 3D ROC CURVES (One-vs-Rest) ====================
    logger.log(f"\n📊 Creating 3D ROC curves for {model_name}...")
    try:
        # Binarize labels for ROC curve
        y_test_bin = label_binarize(y_test, classes=list(range(len(class_names))))
        
        # Create 3D ROC visualization
        fig_roc = go.Figure()
        
        # Store AUC values for each class
        auc_values = []
        
        # Calculate ROC for each class
        for i, class_name in enumerate(class_names):
            if len(np.unique(y_test_bin[:, i])) < 2:
                continue
            
            fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_pred_probs[:, i])
            roc_auc_class = auc(fpr, tpr)
            auc_values.append(roc_auc_class)
            
            # Add 3D curve (using class index as z-axis)
            fig_roc.add_trace(go.Scatter3d(
                x=fpr,
                y=tpr,
                z=[i] * len(fpr),
                mode='lines',
                name=f'{class_name} (AUC={roc_auc_class:.3f})',
                line=dict(width=3),
                hovertemplate=f'Class {class_name}<br>FPR: %{{x:.3f}}<br>TPR: %{{y:.3f}}<extra></extra>'
            ))
        
        # Add diagonal reference plane
        diag_fpr = np.linspace(0, 1, 20)
        diag_tpr = np.linspace(0, 1, 20)
        for i in range(len(class_names)):
            fig_roc.add_trace(go.Scatter3d(
                x=diag_fpr, y=diag_tpr, z=[i] * 20,
                mode='lines',
                line=dict(color='gray', width=1, dash='dash'),
                showlegend=False,
                hoverinfo='skip'
            ))
        
        fig_roc.update_layout(
            title=dict(
                text=f'📈 {model_name} - 3D ROC Curves (One-vs-Rest)',
                font=dict(size=18, color='#333'),
                x=0.5
            ),
            scene=dict(
                xaxis=dict(title='False Positive Rate'),
                yaxis=dict(title='True Positive Rate'),
                zaxis=dict(title='Class', tickvals=list(range(len(class_names))), ticktext=class_names),
                camera=dict(eye=dict(x=1.5, y=-1.5, z=1.0))
            ),
            height=700,
            template='plotly_white',
            legend=dict(font=dict(size=8), x=1.02, y=0.98)
        )
        
        roc_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_roc_curves_3d.png')
        save_and_show_plotly(fig_roc, roc_path, f'{model_name} 3D ROC Curves')
        
    except Exception as e:
        logger.log(f"⚠️ Could not create 3D ROC curves: {e}")
    
    # ==================== 3D PRECISION-RECALL CURVES ====================
    logger.log(f"\n📊 Creating 3D Precision-Recall curves for {model_name}...")
    try:
        fig_pr = go.Figure()
        ap_values = []
        
        for i, class_name in enumerate(class_names):
            if len(np.unique(y_test_bin[:, i])) < 2:
                continue
            
            precision_curve, recall_curve, _ = precision_recall_curve(y_test_bin[:, i], y_pred_probs[:, i])
            ap = average_precision_score(y_test_bin[:, i], y_pred_probs[:, i])
            ap_values.append(ap)
            
            # Add 3D curve
            fig_pr.add_trace(go.Scatter3d(
                x=recall_curve,
                y=precision_curve,
                z=[i] * len(recall_curve),
                mode='lines',
                name=f'{class_name} (AP={ap:.3f})',
                line=dict(width=3),
                hovertemplate=f'Class {class_name}<br>Recall: %{{x:.3f}}<br>Precision: %{{y:.3f}}<extra></extra>'
            ))
        
        fig_pr.update_layout(
            title=dict(
                text=f'📈 {model_name} - 3D Precision-Recall Curves',
                font=dict(size=18, color='#333'),
                x=0.5
            ),
            scene=dict(
                xaxis=dict(title='Recall'),
                yaxis=dict(title='Precision'),
                zaxis=dict(title='Class', tickvals=list(range(len(class_names))), ticktext=class_names),
                camera=dict(eye=dict(x=1.5, y=-1.5, z=1.0))
            ),
            height=700,
            template='plotly_white',
            legend=dict(font=dict(size=8), x=1.02, y=0.98)
        )
        
        pr_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_pr_curves_3d.png')
        save_and_show_plotly(fig_pr, pr_path, f'{model_name} 3D Precision-Recall Curves')
        
    except Exception as e:
        logger.log(f"⚠️ Could not create 3D PR curves: {e}")
    
    # ==================== METRICS DASHBOARD (Visual instead of text) ====================
    logger.log(f"\n📊 Creating visual metrics dashboard for {model_name}...")
    
    # Create comprehensive metrics dashboard
    fig_dash = make_subplots(
        rows=2, cols=3,
        specs=[
            [{"type": "indicator"}, {"type": "indicator"}, {"type": "indicator"}],
            [{"type": "indicator"}, {"type": "indicator"}, {"type": "indicator"}]
        ],
        subplot_titles=['Accuracy', 'Balanced Accuracy', 'F1 Score', 'Cohen\'s Kappa', 'ROC-AUC', 'Matthews CC']
    )
    
    # Row 1: Accuracy metrics
    fig_dash.add_trace(go.Indicator(
        mode="gauge+number",
        value=accuracy * 100,
        domain={'x': [0, 1], 'y': [0, 1]},
        gauge=dict(
            axis=dict(range=[0, 100], ticksuffix='%'),
            bar=dict(color='#2ecc71'),
            steps=[
                dict(range=[0, 50], color='#e74c3c'),
                dict(range=[50, 80], color='#f39c12'),
                dict(range=[80, 100], color='#2ecc71')
            ],
            threshold=dict(line=dict(color="black", width=2), thickness=0.75, value=accuracy*100)
        ),
        number=dict(suffix='%', font=dict(size=24))
    ), row=1, col=1)
    
    fig_dash.add_trace(go.Indicator(
        mode="gauge+number",
        value=balanced_acc * 100,
        gauge=dict(
            axis=dict(range=[0, 100], ticksuffix='%'),
            bar=dict(color='#3498db'),
            steps=[
                dict(range=[0, 50], color='#e74c3c'),
                dict(range=[50, 80], color='#f39c12'),
                dict(range=[80, 100], color='#2ecc71')
            ]
        ),
        number=dict(suffix='%', font=dict(size=24))
    ), row=1, col=2)
    
    fig_dash.add_trace(go.Indicator(
        mode="gauge+number",
        value=f1 * 100,
        gauge=dict(
            axis=dict(range=[0, 100], ticksuffix='%'),
            bar=dict(color='#9b59b6'),
            steps=[
                dict(range=[0, 50], color='#e74c3c'),
                dict(range=[50, 80], color='#f39c12'),
                dict(range=[80, 100], color='#2ecc71')
            ]
        ),
        number=dict(suffix='%', font=dict(size=24))
    ), row=1, col=3)
    
    # Row 2: Advanced metrics
    fig_dash.add_trace(go.Indicator(
        mode="gauge+number",
        value=cohen_kappa,
        gauge=dict(
            axis=dict(range=[-1, 1]),
            bar=dict(color='#e67e22'),
            steps=[
                dict(range=[-1, 0], color='#e74c3c'),
                dict(range=[0, 0.6], color='#f39c12'),
                dict(range=[0.6, 1], color='#2ecc71')
            ]
        ),
        number=dict(font=dict(size=24))
    ), row=2, col=1)
    
    fig_dash.add_trace(go.Indicator(
        mode="gauge+number",
        value=(roc_auc or 0) * 100,
        gauge=dict(
            axis=dict(range=[0, 100], ticksuffix='%'),
            bar=dict(color='#1abc9c'),
            steps=[
                dict(range=[0, 50], color='#e74c3c'),
                dict(range=[50, 80], color='#f39c12'),
                dict(range=[80, 100], color='#2ecc71')
            ]
        ),
        number=dict(suffix='%', font=dict(size=24))
    ), row=2, col=2)
    
    fig_dash.add_trace(go.Indicator(
        mode="gauge+number",
        value=mcc,
        gauge=dict(
            axis=dict(range=[-1, 1]),
            bar=dict(color='#34495e'),
            steps=[
                dict(range=[-1, 0], color='#e74c3c'),
                dict(range=[0, 0.5], color='#f39c12'),
                dict(range=[0.5, 1], color='#2ecc71')
            ]
        ),
        number=dict(font=dict(size=24))
    ), row=2, col=3)
    
    fig_dash.update_layout(
        title=dict(
            text=f'📊 {model_name} - Metrics Dashboard',
            font=dict(size=22, color='#333'),
            x=0.5
        ),
        height=600,
        template='plotly_white'
    )
    
    dash_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_metrics_dashboard.png')
    save_and_show_plotly(fig_dash, dash_path, f'{model_name} Metrics Dashboard')
    
    # ==================== PER-CLASS METRICS RADAR CHART ====================
    logger.log(f"\n📊 Creating per-class metrics visualization for {model_name}...")
    
    # Create radar chart for per-class F1 scores
    fig_radar = go.Figure()
    
    # Add F1 scores as radar
    fig_radar.add_trace(go.Scatterpolar(
        r=list(f1_per_class) + [f1_per_class[0]],  # Close the polygon
        theta=list(class_names) + [class_names[0]],
        fill='toself',
        name='F1 Score',
        line=dict(color='#3498db', width=2),
        fillcolor='rgba(52, 152, 219, 0.3)'
    ))
    
    # Add precision
    fig_radar.add_trace(go.Scatterpolar(
        r=list(precision_per_class) + [precision_per_class[0]],
        theta=list(class_names) + [class_names[0]],
        fill='toself',
        name='Precision',
        line=dict(color='#2ecc71', width=2),
        fillcolor='rgba(46, 204, 113, 0.2)'
    ))
    
    # Add recall
    fig_radar.add_trace(go.Scatterpolar(
        r=list(recall_per_class) + [recall_per_class[0]],
        theta=list(class_names) + [class_names[0]],
        fill='toself',
        name='Recall',
        line=dict(color='#e74c3c', width=2),
        fillcolor='rgba(231, 76, 60, 0.2)'
    ))
    
    fig_radar.update_layout(
        polar=dict(
            radialaxis=dict(visible=True, range=[0, 1])
        ),
        showlegend=True,
        title=dict(
            text=f'🎯 {model_name} - Per-Class Metrics Radar',
            font=dict(size=18, color='#333'),
            x=0.5
        ),
        height=600,
        template='plotly_white'
    )
    
    radar_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_per_class_radar.png')
    save_and_show_plotly(fig_radar, radar_path, f'{model_name} Per-Class Radar')
    
    # ==================== 3D CONFIDENCE DISTRIBUTION ====================
    logger.log(f"\n📊 Creating 3D confidence distribution for {model_name}...")
    
    # Create fully 3D confidence visualization
    fig_conf = go.Figure()
    
    # 3D scatter: True class × Predicted class × Confidence
    # Correct predictions (green)
    correct_indices = np.where(correct_mask)[0]
    if len(correct_indices) > 0:
        fig_conf.add_trace(go.Scatter3d(
            x=y_test[correct_indices],
            y=y_pred[correct_indices],
            z=pred_confidences[correct_indices],
            mode='markers',
            name='Correct',
            marker=dict(
                size=4,
                color='#2ecc71',
                opacity=0.6,
                symbol='circle'
            ),
            hovertemplate='True: %{x}<br>Pred: %{y}<br>Conf: %{z:.3f}<extra></extra>'
        ))
    
    # Wrong predictions (red)
    wrong_indices = np.where(~correct_mask)[0]
    if len(wrong_indices) > 0:
        fig_conf.add_trace(go.Scatter3d(
            x=y_test[wrong_indices],
            y=y_pred[wrong_indices],
            z=pred_confidences[wrong_indices],
            mode='markers',
            name='Wrong',
            marker=dict(
                size=5,
                color='#e74c3c',
                opacity=0.8,
                symbol='x'
            ),
            hovertemplate='True: %{x}<br>Pred: %{y}<br>Conf: %{z:.3f}<extra></extra>'
        ))
    
    # Add diagonal plane (where predictions are correct)
    diagonal_x = list(range(n_classes))
    diagonal_y = list(range(n_classes))
    fig_conf.add_trace(go.Scatter3d(
        x=diagonal_x,
        y=diagonal_y,
        z=[1.0] * n_classes,
        mode='lines',
        name='Perfect (diagonal)',
        line=dict(color='gold', width=5),
        hoverinfo='skip'
    ))
    
    fig_conf.update_layout(
        title=dict(
            text=f'🎲 {model_name} - 3D Confidence Analysis (Green=Correct, Red=Wrong)',
            font=dict(size=18, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(title='True Class', tickvals=list(range(n_classes)), ticktext=class_names),
            yaxis=dict(title='Predicted Class', tickvals=list(range(n_classes)), ticktext=class_names),
            zaxis=dict(title='Confidence', range=[0, 1]),
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.0))
        ),
        height=700,
        template='plotly_white'
    )
    
    conf_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_confidence_analysis.png')
    save_and_show_plotly(fig_conf, conf_path, f'{model_name} 3D Confidence Analysis')
    
    # ==================== CUMULATIVE GAIN & LIFT CHARTS ====================
    logger.log(f"\n📊 Creating Cumulative Gain and Lift charts for {model_name}...")
    try:
        # Create figure with subplots
        fig_gain_lift = make_subplots(
            rows=1, cols=2,
            subplot_titles=['📈 Cumulative Gain Chart', '📊 Lift Chart'],
            horizontal_spacing=0.12
        )
        
        # Calculate for each class (one-vs-rest)
        n_samples = len(y_test)
        colors = ['#e74c3c', '#3498db', '#2ecc71', '#9b59b6', '#f39c12', '#1abc9c',
                  '#e67e22', '#34495e', '#16a085', '#c0392b', '#2980b9', '#27ae60',
                  '#8e44ad', '#d35400', '#7f8c8d', '#f1c40f', '#e91e63', '#00bcd4',
                  '#795548', '#607d8b', '#ff5722', '#4caf50', '#673ab7', '#009688',
                  '#ff9800', '#03a9f4']
        
        for class_idx, class_name in enumerate(class_names[:10]):  # Top 10 classes for readability
            # Get probabilities for this class
            class_probs = y_pred_probs[:, class_idx]
            y_true_binary = (y_test == class_idx).astype(int)
            
            # Skip if no positive samples
            if y_true_binary.sum() == 0:
                continue
            
            # Sort by predicted probability descending
            sorted_indices = np.argsort(class_probs)[::-1]
            y_true_sorted = y_true_binary[sorted_indices]
            
            # Calculate cumulative gain
            cumulative_gains = np.cumsum(y_true_sorted) / y_true_binary.sum()
            percentile = np.arange(1, len(y_true_sorted) + 1) / len(y_true_sorted)
            
            # Add gain curve
            fig_gain_lift.add_trace(go.Scatter(
                x=percentile * 100,
                y=cumulative_gains * 100,
                mode='lines',
                name=f'{class_name}',
                line=dict(color=colors[class_idx % len(colors)], width=2),
                hovertemplate=f'{class_name}<br>%{{x:.1f}}% samples<br>%{{y:.1f}}% gain<extra></extra>'
            ), row=1, col=1)
            
            # Calculate lift
            # Lift = (Cumulative True Positives / Total Sampled) / (Total Positives / Total Samples)
            expected_rate = y_true_binary.mean()
            if expected_rate > 0:
                lift = (np.cumsum(y_true_sorted) / np.arange(1, len(y_true_sorted) + 1)) / expected_rate
                
                fig_gain_lift.add_trace(go.Scatter(
                    x=percentile * 100,
                    y=lift,
                    mode='lines',
                    name=f'{class_name}',
                    line=dict(color=colors[class_idx % len(colors)], width=2),
                    showlegend=False,
                    hovertemplate=f'{class_name}<br>%{{x:.1f}}% samples<br>Lift: %{{y:.2f}}x<extra></extra>'
                ), row=1, col=2)
        
        # Add baseline (random model) to gain chart
        fig_gain_lift.add_trace(go.Scatter(
            x=[0, 100],
            y=[0, 100],
            mode='lines',
            name='Random',
            line=dict(color='gray', width=2, dash='dash'),
            hovertemplate='Random Model<extra></extra>'
        ), row=1, col=1)
        
        # Add baseline lift = 1
        fig_gain_lift.add_trace(go.Scatter(
            x=[0, 100],
            y=[1, 1],
            mode='lines',
            name='Baseline',
            line=dict(color='gray', width=2, dash='dash'),
            showlegend=False,
            hovertemplate='Baseline Lift = 1<extra></extra>'
        ), row=1, col=2)
        
        fig_gain_lift.update_layout(
            title=dict(
                text=f'📊 {model_name} - Cumulative Gain & Lift Analysis (Top 10 Classes)',
                font=dict(size=18, color='#333'),
                x=0.5
            ),
            height=500,
            template='plotly_white',
            legend=dict(font=dict(size=9), orientation='h', y=-0.15)
        )
        
        fig_gain_lift.update_xaxes(title_text='Percentage of Samples (%)', row=1, col=1)
        fig_gain_lift.update_yaxes(title_text='Cumulative Gain (%)', row=1, col=1)
        fig_gain_lift.update_xaxes(title_text='Percentage of Samples (%)', row=1, col=2)
        fig_gain_lift.update_yaxes(title_text='Lift', row=1, col=2)
        
        gain_lift_path = os.path.join(CONFIG['metrics_path'], f'{model_name}_gain_lift_charts.png')
        save_and_show_plotly(fig_gain_lift, gain_lift_path, f'{model_name} Gain & Lift Charts')
        
    except Exception as e:
        logger.log(f"⚠️ Could not create Gain/Lift charts: {e}")
    
    # Return comprehensive metrics
    return {
        'accuracy': accuracy,
        'balanced_accuracy': balanced_acc,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'f1_macro': f1_macro,
        'cohen_kappa': cohen_kappa,
        'matthews_corrcoef': mcc,
        'roc_auc': roc_auc,
        'log_loss': logloss,
        'top3_accuracy': top3_acc,
        'top5_accuracy': top5_acc,
        'avg_confidence': avg_confidence,
        'avg_conf_correct': avg_conf_correct,
        'avg_conf_wrong': avg_conf_wrong
    }

# ==================== FINAL COMPARISON ====================
def plot_model_comparison(results):
    """Plot final model comparison with SOLID 3D bars"""
    
    logger.section("MODEL COMPARISON")
    
    # Filter valid results
    valid_results = {k: v for k, v in results.items() if v is not None}
    
    if not valid_results:
        logger.log("⚠️ No valid results to compare")
        return
    
    # Create comparison dataframe
    df = pd.DataFrame(valid_results).T
    df.index.name = 'Model'
    
    logger.log("\n📊 Final Results:")
    logger.log(df.to_string())
    
    # Find best model
    best_model = df['accuracy'].idxmax()
    best_acc = df.loc[best_model, 'accuracy']
    logger.log(f"\n🏆 Best Model: {best_model} ({best_acc:.4f})")
    
    # Save results
    results_path = os.path.join(CONFIG['metrics_path'], 'final_results.csv')
    df.to_csv(results_path)
    logger.log(f"💾 Results saved to: {results_path}")
    
    # Create 3D model comparison visualization with SOLID bars
    models = list(valid_results.keys())
    n_models = len(models)
    metrics = ['accuracy', 'precision', 'recall', 'f1_score', 'cohen_kappa']
    
    fig = go.Figure()
    
    # Create SOLID 3D bars for each model and metric
    metric_colors = ['#2ecc71', '#3498db', '#e74c3c', '#9b59b6', '#f39c12']
    
    for m_idx, metric in enumerate(metrics):
        values = [v.get(metric, 0) or 0 for v in valid_results.values()]
        
        for i, (model, val) in enumerate(zip(models, values)):
            # Create solid 3D bar with proper position
            x_center = i + (m_idx - 2) * 0.18  # Spread metrics within model group
            bar_width = 0.15
            bar_depth = 0.4
            
            bar = create_3d_bar_mesh(
                x_center=x_center,
                y_center=0,
                z_height=val,
                bar_width=bar_width,
                bar_depth=bar_depth,
                color=metric_colors[m_idx],
                opacity=0.95,
                name=metric.replace('_', ' ').title() if i == 0 else '',
                hovertemplate=f'<b>{model}</b><br>{metric}: {val:.4f}<extra></extra>'
            )
            # Set legend group manually
            bar.update(showlegend=(i == 0), legendgroup=metric)
            fig.add_trace(bar)
    
    # Add text labels
    fig.add_trace(go.Scatter3d(
        x=list(range(n_models)),
        y=[0] * n_models,
        z=[1.08] * n_models,
        mode='text',
        text=models,
        textfont=dict(size=12, color='black', family='Arial Black'),
        hoverinfo='skip',
        showlegend=False
    ))
    
    fig.update_layout(
        title=dict(
            text='🏆 3D Model Comparison (Rotate to explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(title='Model', tickvals=list(range(n_models)), ticktext=models),
            yaxis=dict(title='', showticklabels=False, range=[-0.8, 0.8]),
            zaxis=dict(title='Score', range=[0, 1.15]),
            camera=dict(eye=dict(x=1.5, y=1.5, z=0.8)),
            bgcolor='rgba(250,250,250,0.9)'
        ),
        height=650,
        template='plotly_white',
        legend=dict(
            title=dict(text='Metrics'),
            yanchor="top",
            y=0.99,
            xanchor="left",
            x=1.02
        )
    )
    
    comparison_path = os.path.join(CONFIG['metrics_path'], 'model_comparison.png')
    save_and_show_plotly(fig, comparison_path, '3D Model Comparison')

def plot_model_comparison_3d(results):
    """Create INTERACTIVE 3D visualization comparing all models with Plotly"""
    
    logger.log("\n📊 Creating interactive 3D model comparison...")
    
    # Filter valid results
    valid_results = {k: v for k, v in results.items() if v is not None}
    
    if len(valid_results) < 2:
        logger.log("⚠️ Need at least 2 models for 3D comparison")
        return
    
    models = list(valid_results.keys())
    metrics = ['accuracy', 'precision', 'recall', 'f1_score']
    
    # Create 3D scatter showing all metrics for each model
    fig = go.Figure()
    
    colors = px.colors.qualitative.Set1
    
    for i, model in enumerate(models):
        values = [valid_results[model].get(m, 0) or 0 for m in metrics]
        
        # Create 3D representation: X = metric index, Y = model index, Z = score
        fig.add_trace(go.Scatter3d(
            x=list(range(len(metrics))),
            y=[i] * len(metrics),
            z=values,
            mode='markers+lines',
            name=model,
            marker=dict(
                size=12,
                color=colors[i % len(colors)],
                opacity=0.8,
                line=dict(width=2, color='darkgray')
            ),
            line=dict(color=colors[i % len(colors)], width=3),
            hovertemplate=f'<b>{model}</b><br>%{{text}}: %{{z:.4f}}<extra></extra>',
            text=[m.replace('_', ' ').title() for m in metrics]
        ))
    
    fig.update_layout(
        title=dict(
            text='🏆 3D Model Comparison (Rotate & Explore!)',
            font=dict(size=20, color='#333'),
            x=0.5
        ),
        scene=dict(
            xaxis=dict(title='Metric', tickvals=list(range(len(metrics))), 
                      ticktext=[m.replace('_', ' ').title() for m in metrics]),
            yaxis=dict(title='Model', tickvals=list(range(len(models))), ticktext=models),
            zaxis=dict(title='Score', range=[0.8, 1.0]),
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.2))
        ),
        height=700,
        template='plotly_white'
    )
    
    save_path = os.path.join(CONFIG['metrics_path'], 'model_comparison_3d.png')
    save_and_show_plotly(fig, save_path, '3D Model Comparison')

# ==================== MAIN PIPELINE ====================
def main():
    """Main training pipeline
    
    FLOW:
    1. Collect and store all dataset information
    2. Process dataset with MediaPipe
    3. Collect all model information (architecture, params, status)
    4. Test existing models BEFORE training (baseline metrics)
    5. Train models (resume or from scratch)
    6. Evaluate trained models
    7. Compare results (baseline vs trained)
    """
    
    start_time = time.time()
    
    print("\n" + "=" * 80)
    print("  ASL ALPHABET CLASSIFICATION - UNIFIED TRAINING PIPELINE")
    print("=" * 80)
    print(f"  Start time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"  GPU Available: {HAS_GPU}")
    print(f"  TensorFlow: {tf.__version__}")
    print("=" * 80 + "\n")
    
    # ==================== PHASE 1: DATASET INFO ====================
    logger.section("PHASE 1: DATASET INFORMATION")
    logger.log("Collecting comprehensive dataset information...")
    
    metrics = analyze_dataset()
    
    if metrics['total_classes'] == 0:
        logger.log("❌ No valid classes found. Please check CONFIG paths.")
        return
    
    # ==================== PHASE 2: MEDIAPIPE PROCESSING ====================
    logger.section("PHASE 2: MEDIAPIPE PROCESSING")
    
    class_names = process_dataset_with_mediapipe()
    
    if len(class_names) == 0:
        logger.log("❌ No classes after processing. Please check dataset.")
        return
    
    num_classes = len(class_names)
    
    # Save class names
    class_names_path = os.path.join(CONFIG['models_path'], 'class_names.json')
    with open(class_names_path, 'w') as f:
        json.dump(class_names, f, indent=2)
    logger.log(f"💾 Class names saved to: {class_names_path}")
    
    # ==================== PHASE 3: MODEL INFO COLLECTION ====================
    logger.section("PHASE 3: MODEL INFORMATION")
    logger.log("Collecting comprehensive model information...")
    
    model_info = collect_model_info(num_classes)
    
    # ==================== PHASE 4: LOAD DATASET ====================
    logger.section("PHASE 4: LOADING DATASET")
    
    train_paths, train_labels = load_dataset(CONFIG['cropped_train_path'], class_names)
    test_paths, test_labels = load_dataset(CONFIG['cropped_test_path'], class_names)
    
    logger.log(f"📊 Training samples: {len(train_paths)}")
    logger.log(f"📊 Test samples: {len(test_paths)}")
    logger.log(f"📊 Classes: {num_classes}")
    
    if len(train_paths) == 0:
        logger.log("❌ No training data found.")
        return
    
    # ==================== PHASE 5: BASELINE TESTING ====================
    if len(test_paths) > 0:
        logger.section("PHASE 5: PRE-TRAINING BASELINE TEST")
        logger.log("Testing existing models to get baseline metrics...")
        
        baseline_results = test_models_before_training(
            test_paths, test_labels, class_names, model_info
        )
    else:
        logger.log("⚠️ No test data available for baseline testing")
        baseline_results = {}
    
    # ==================== PHASE 6: TRAIN/VAL SPLIT ====================
    logger.section("PHASE 6: PREPARING TRAINING DATA")
    
    # Split train/val
    try:
        train_paths, val_paths, train_labels, val_labels = train_test_split(
            train_paths, train_labels,
            test_size=CONFIG['val_split'],
            random_state=42,
            stratify=train_labels
        )
    except Exception as e:
        logger.log(f"⚠️ Stratified split failed: {e}. Using random split.")
        split_idx = int(len(train_paths) * (1 - CONFIG['val_split']))
        train_paths, val_paths = train_paths[:split_idx], train_paths[split_idx:]
        train_labels, val_labels = train_labels[:split_idx], train_labels[split_idx:]
    
    logger.log(f"📊 Training samples: {len(train_paths)}")
    logger.log(f"📊 Validation samples: {len(val_paths)}")
    logger.log(f"📊 Test samples: {len(test_paths)}")
    
    # ==================== PHASE 7: INTELLIGENT TRAINING DECISIONS ====================
    logger.section("PHASE 7: INTELLIGENT TRAINING ANALYSIS")
    
    results = {}
    
    # ===== INTELLIGENT TRAINING DECISION THRESHOLDS =====
    SKIP_THRESHOLD = 0.98       # If accuracy >= 98%, skip training (model is excellent)
    GOOD_THRESHOLD = 0.95       # If accuracy >= 95%, model is good but can be fine-tuned
    MODERATE_THRESHOLD = 0.30   # If accuracy >= 30%, resume training
    # Below 30% = retrain from scratch (model incompatible)
    
    def analyze_model_decision(model_name, baseline):
        """
        Intelligent analysis to decide training action for each model.
        
        Returns: (action, reason, details)
        - action: 'SKIP', 'FINE_TUNE', 'RESUME', 'SCRATCH'
        - reason: Human-readable explanation
        - details: Dict with metrics
        """
        if not baseline.get('baseline_tested'):
            reason = baseline.get('reason', 'No existing model')
            return 'SCRATCH', f"No existing model found: {reason}", {'reason': reason}
        
        accuracy = baseline.get('accuracy', 0)
        kappa = baseline.get('cohen_kappa', 0)
        f1 = baseline.get('f1_score', 0)
        balanced_acc = baseline.get('balanced_accuracy', accuracy)
        
        details = {
            'accuracy': accuracy,
            'cohen_kappa': kappa,
            'f1_score': f1,
            'balanced_accuracy': balanced_acc
        }
        
        # Decision tree based on multiple metrics
        if accuracy >= SKIP_THRESHOLD and kappa >= 0.97:
            return 'SKIP', f"Excellent performance (Acc: {accuracy*100:.2f}%, Kappa: {kappa:.4f})", details
        
        if accuracy >= GOOD_THRESHOLD and kappa >= 0.94:
            return 'FINE_TUNE', f"Good performance, light fine-tuning recommended (Acc: {accuracy*100:.2f}%)", details
        
        if accuracy >= MODERATE_THRESHOLD:
            return 'RESUME', f"Moderate performance, continue training (Acc: {accuracy*100:.2f}%)", details
        
        # Check for preprocessing mismatch (high confidence but low accuracy)
        avg_conf = baseline.get('avg_confidence', 0)
        if accuracy < MODERATE_THRESHOLD and avg_conf > 0.8:
            return 'SCRATCH', f"Preprocessing mismatch detected (Acc: {accuracy*100:.2f}%, Conf: {avg_conf*100:.1f}%)", details
        
        return 'SCRATCH', f"Low performance, retrain from scratch (Acc: {accuracy*100:.2f}%)", details
    
    # Analyze all models and create training plan
    training_plan = {}
    for model_name in CONFIG['train_models']:
        baseline = baseline_results.get(model_name, {})
        action, reason, details = analyze_model_decision(model_name, baseline)
        training_plan[model_name] = {
            'action': action,
            'reason': reason,
            'details': details,
            'baseline': baseline
        }
    
    # Display intelligent training plan
    logger.log("\n" + "=" * 70)
    logger.log("🧠 INTELLIGENT TRAINING DECISION SUMMARY")
    logger.log("=" * 70)
    logger.log(f"\n   Thresholds:")
    logger.log(f"   • SKIP training if accuracy ≥ {SKIP_THRESHOLD*100:.0f}% (excellent)")
    logger.log(f"   • FINE-TUNE if accuracy ≥ {GOOD_THRESHOLD*100:.0f}% (good)")
    logger.log(f"   • RESUME if accuracy ≥ {MODERATE_THRESHOLD*100:.0f}% (moderate)")
    logger.log(f"   • SCRATCH if accuracy < {MODERATE_THRESHOLD*100:.0f}% (incompatible)")
    
    skip_count = sum(1 for p in training_plan.values() if p['action'] == 'SKIP')
    fine_tune_count = sum(1 for p in training_plan.values() if p['action'] == 'FINE_TUNE')
    resume_count = sum(1 for p in training_plan.values() if p['action'] == 'RESUME')
    scratch_count = sum(1 for p in training_plan.values() if p['action'] == 'SCRATCH')
    
    logger.log(f"\n   📊 Training Plan Summary:")
    logger.log(f"   • SKIP (no training needed): {skip_count} models")
    logger.log(f"   • FINE-TUNE (light training): {fine_tune_count} models")
    logger.log(f"   • RESUME (continue training): {resume_count} models")
    logger.log(f"   • SCRATCH (full retrain): {scratch_count} models")
    
    logger.log(f"\n   📋 Per-Model Decisions:")
    logger.log(f"   {'-' * 66}")
    
    action_icons = {'SKIP': '⏭️ ', 'FINE_TUNE': '🔧', 'RESUME': '🔄', 'SCRATCH': '🆕'}
    action_colors = {'SKIP': '🟢', 'FINE_TUNE': '🟡', 'RESUME': '🟠', 'SCRATCH': '🔴'}
    
    for model_name, plan in training_plan.items():
        icon = action_icons[plan['action']]
        color = action_colors[plan['action']]
        acc = plan['details'].get('accuracy', 0) * 100
        logger.log(f"   {color} {icon} {model_name}: {plan['action']}")
        logger.log(f"      └─ {plan['reason']}")
    
    logger.log(f"   {'-' * 66}")
    logger.log("=" * 70 + "\n")
    
    # ===== PRIORITY TRAINING: Sort models by action priority =====
    def get_training_priority(model_name):
        """Return priority score - lower = train first, -1 = skip"""
        plan = training_plan.get(model_name, {})
        action = plan.get('action', 'SCRATCH')
        
        if action == 'SKIP':
            return 99  # Will be filtered out
        elif action == 'SCRATCH':
            return 0   # Highest priority - train first
        elif action == 'FINE_TUNE':
            return 2   # Lower priority
        elif action == 'RESUME':
            return 1   # Medium priority
        return 3
    
    # Filter and sort models
    models_to_train = [m for m in CONFIG['train_models'] if training_plan[m]['action'] != 'SKIP']
    sorted_models = sorted(models_to_train, key=get_training_priority)
    
    # Handle skipped models - add their baseline results directly
    for model_name in CONFIG['train_models']:
        if training_plan[model_name]['action'] == 'SKIP':
            baseline = training_plan[model_name]['baseline']
            logger.log(f"\n⏭️  SKIPPING {model_name} - Already excellent!")
            logger.log(f"   Accuracy: {baseline.get('accuracy', 0)*100:.2f}%")
            logger.log(f"   Cohen's Kappa: {baseline.get('cohen_kappa', 0):.4f}")
            logger.log(f"   Using baseline results as final results.")
            
            # Use baseline as final results
            results[model_name] = {
                'accuracy': baseline.get('accuracy', 0),
                'balanced_accuracy': baseline.get('balanced_accuracy', 0),
                'precision': baseline.get('precision', 0),
                'recall': baseline.get('recall', 0),
                'f1_score': baseline.get('f1_score', 0),
                'f1_macro': baseline.get('f1_macro', 0),
                'cohen_kappa': baseline.get('cohen_kappa', 0),
                'matthews_corrcoef': baseline.get('matthews_corrcoef', 0),
                'roc_auc': baseline.get('roc_auc', None),
                'log_loss': baseline.get('log_loss', None),
                'top3_accuracy': baseline.get('top3_accuracy', None),
                'top5_accuracy': baseline.get('top5_accuracy', None),
                'avg_confidence': baseline.get('avg_confidence', 0),
                'skipped_training': True,
                'reason': 'Excellent baseline performance'
            }
    
    if not sorted_models:
        logger.log("\n🎉 ALL MODELS HAVE EXCELLENT PERFORMANCE!")
        logger.log("   No training required. Using baseline results.")
    else:
        logger.log(f"\n📋 TRAINING ORDER ({len(sorted_models)} models to train):")
        logger.log("=" * 50)
        for i, model_name in enumerate(sorted_models, 1):
            plan = training_plan[model_name]
            icon = action_icons[plan['action']]
            acc = plan['details'].get('accuracy', 0) * 100
            logger.log(f"  {i}. {icon} {model_name}: {plan['action']} (baseline: {acc:.1f}%)")
        logger.log("=" * 50 + "\n")
    
    for model_name in sorted_models:
        if model_name not in MODEL_BUILDERS:
            logger.log(f"⚠️ Unknown model: {model_name}")
            continue
        
        plan = training_plan[model_name]
        action = plan['action']
        baseline = plan['baseline']
        baseline_accuracy = baseline.get('accuracy', 0) if baseline.get('baseline_tested') else 0
        
        # Determine training parameters based on action
        force_scratch = (action == 'SCRATCH')
        
        # Print clear training mode with box
        logger.log(f"\n" + "="*60)
        if action == 'SCRATCH':
            logger.log(f"🆕 {model_name}: Training from SCRATCH")
            logger.log(f"   Reason: {plan['reason']}")
        elif action == 'FINE_TUNE':
            logger.log(f"🔧 {model_name}: FINE-TUNING (fewer epochs, lower LR)")
            logger.log(f"   Baseline accuracy: {baseline_accuracy*100:.2f}%")
            logger.log(f"   Reason: {plan['reason']}")
        elif action == 'RESUME':
            logger.log(f"🔄 {model_name}: RESUMING training")
            logger.log(f"   Baseline accuracy: {baseline_accuracy*100:.2f}%")
            logger.log(f"   Reason: {plan['reason']}")
        logger.log("="*60)
        
        try:
            # Train (pass training_mode for intelligent training)
            model, history = train_model(
                model_name, train_paths, train_labels,
                val_paths, val_labels, num_classes, class_names,
                force_scratch=False,
                training_mode=action  # Pass the intelligent decision: SCRATCH, RESUME, FINE_TUNE
            )
            
            # Evaluate
            model_results = evaluate_model(
                model, model_name, test_paths, test_labels, class_names
            )
            
            results[model_name] = model_results
            
            # Generate 3D embeddings visualization
            try:
                visualize_feature_embeddings_3d(
                    model, model_name, test_paths, test_labels, class_names
                )
            except Exception as e:
                logger.log(f"⚠️ 3D embeddings visualization failed: {e}")
            
            # Compare with baseline
            if model_name in baseline_results and baseline_results[model_name].get('baseline_tested'):
                baseline_acc = baseline_results[model_name]['accuracy']
                trained_acc = model_results['accuracy']
                improvement = (trained_acc - baseline_acc) * 100
                logger.log(f"\n📈 Improvement for {model_name}:")
                logger.log(f"   Baseline: {baseline_acc*100:.2f}% → Trained: {trained_acc*100:.2f}%")
                logger.log(f"   Change: {improvement:+.2f}%")
            
            # Clear memory
            del model
            keras.backend.clear_session()
            gc.collect()
            time.sleep(3)
            
            logger.log(f"\n✅ {model_name} complete. Memory cleared.\n")
            
        except Exception as e:
            logger.log(f"❌ {model_name} failed: {e}")
            import traceback
            logger.log(traceback.format_exc())
            results[model_name] = None
    
    # ==================== PHASE 8: FINAL COMPARISON ====================
    logger.section("PHASE 8: FINAL COMPARISON")
    
    plot_model_comparison(results)
    plot_model_comparison_3d(results)
    
    # Compare baseline vs trained
    compare_baseline_vs_trained(baseline_results, results)
    
    # ==================== SUMMARY ====================
    total_time = time.time() - start_time
    
    logger.section("TRAINING COMPLETE")
    logger.log(f"⏱️ Total time: {total_time/60:.1f} minutes ({total_time/3600:.2f} hours)")
    logger.log(f"💾 Models saved to: {CONFIG['models_path']}")
    logger.log(f"📊 Metrics saved to: {CONFIG['metrics_path']}")
    logger.log(f"📋 Log saved to: {logger.log_file}")
    
    # Print summary of all saved files
    logger.log(f"\n📁 OUTPUT FILES SUMMARY:")
    logger.log(f"-" * 50)
    if os.path.exists(CONFIG['metrics_path']):
        for f in sorted(os.listdir(CONFIG['metrics_path'])):
            filepath = os.path.join(CONFIG['metrics_path'], f)
            size = os.path.getsize(filepath) / 1024  # KB
            logger.log(f"  📄 {f} ({size:.1f} KB)")
    
    if os.path.exists(CONFIG['models_path']):
        logger.log(f"\n📁 SAVED MODELS:")
        logger.log(f"-" * 50)
        for f in sorted(os.listdir(CONFIG['models_path'])):
            if f.endswith('.keras'):
                filepath = os.path.join(CONFIG['models_path'], f)
                size = os.path.getsize(filepath) / (1024 * 1024)  # MB
                logger.log(f"  🧠 {f} ({size:.1f} MB)")
    
    print("\n" + "=" * 80)
    print("  ✅ TRAINING PIPELINE COMPLETE!")
    print("=" * 80 + "\n")

def compare_baseline_vs_trained(baseline_results, trained_results):
    """Compare baseline metrics vs trained metrics"""
    
    logger.log("\n📊 BASELINE vs TRAINED COMPARISON:")
    logger.log("=" * 80)
    logger.log(f"{'Model':<20} {'Baseline':>12} {'Trained':>12} {'Change':>12}")
    logger.log("-" * 80)
    
    for model_name in CONFIG['train_models']:
        baseline = baseline_results.get(model_name, {})
        trained = trained_results.get(model_name, {})
        
        if baseline.get('baseline_tested') and trained:
            baseline_acc = baseline['accuracy'] * 100
            trained_acc = trained['accuracy'] * 100
            change = trained_acc - baseline_acc
            change_str = f"{change:+.2f}%" if change >= 0 else f"{change:.2f}%"
            logger.log(f"{model_name:<20} {baseline_acc:>11.2f}% {trained_acc:>11.2f}% {change_str:>12}")
        elif trained:
            trained_acc = trained['accuracy'] * 100
            logger.log(f"{model_name:<20} {'N/A':>12} {trained_acc:>11.2f}% {'NEW':>12}")
        else:
            logger.log(f"{model_name:<20} {'N/A':>12} {'FAILED':>12} {'N/A':>12}")
    
    logger.log("=" * 80)
    
    # Save comparison
    comparison = {
        'baseline': baseline_results,
        'trained': {k: v for k, v in trained_results.items() if v is not None}
    }
    
    comparison_path = os.path.join(CONFIG['metrics_path'], 'baseline_vs_trained.json')
    with open(comparison_path, 'w') as f:
        json.dump(comparison, f, indent=2, default=str)
    
    logger.log(f"💾 Comparison saved to: {comparison_path}")

# ==================== RUN ====================
if __name__ == "__main__":
    main()
```
